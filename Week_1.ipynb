## 1️⃣ Introduction
Modern Database Management Systems (**DBMS**) are highly sophisticated software platforms that underpin almost every sector of the data-driven economy. From **ERP** to **e-commerce**, **analytics**, and **scientific research**, they ensure data is stored, retrieved, and processed efficiently and reliably.

## 2️⃣ DBMS Anatomy: Processes, Memory, and I/O

### 2.1 Process-Level Architecture
- **Query Processor** — Parses SQL and generates optimized execution plans.
- **Transaction Manager** — Ensures ACID compliance.
- **Concurrency Control Manager** — Manages locking, timestamps, MVCC.
- **Recovery Manager** — Restores consistent state after crashes.
- **Storage Manager** — Reads/writes pages, manages free space, interacts with storage.

### 2.2 Memory Architecture
- **Buffer Pool** — Caches disk pages.
- **Log Buffer** — Holds transaction logs before flush.
- **Sort/Join Buffers** — Temporary storage during operations.
- **Catalog Cache** — Stores frequently accessed metadata.

### 2.3 I/O Subsystem
- Page-oriented I/O (4–16 KB).
- Minimizing random I/O is key to performance.

---

## 3️⃣ The Storage Manager

### 3.1 Responsibilities
- Allocate/deallocate disk space.
- Map logical records to physical locations.
- Manage free space.
- Coordinate with buffer pool.
- Ensure durability.

### 3.2 Subcomponents (Elmasri & Navathe)
- **File Manager** — Organizes data files & space allocation.
- **Buffer Manager** — Controls cached pages.
- **Record Manager** — Handles record layout & CRUD operations.

---

## 4️⃣ Record Formats

### Fixed-Length Records
- Same space per record.
- Easy offset calculations.
- Wastes space if fields are sparse.

### Variable-Length Records
- Space-efficient.
- Requires offsets/length indicators.

### Alignment & Padding
- Align to word boundaries for CPU efficiency (can cause fragmentation).

---

## 5️⃣ Slotted Pages

**Structure:**
- **Page Header** — Metadata (slot count, free space start, type).
- **Slot Directory** — Offset & length per record.
- **Record Area** — Records grow backward, slots grow forward.

**Benefits:**
- Efficient variable-length record management.
- Stable slot references.
- Supports compaction.

---

## 6️⃣ Buffer Pool Design

### Operation
- On **hit** — return page from memory.
- On **miss** — load from disk, possibly evict another page.

### Replacement Policies
- **LRU** — Evict least recently used.
- **MRU** — Evict most recently used.
- **Clock** — Approximate LRU.
- **Priority-based** — Keep critical pages.

### Dirty Pages
- **Steal** — Evict dirty pages before commit.
- **No-Steal** — Keep dirty pages until commit.

---

## 7️⃣ Catalogs, Metadata, and Data Dictionaries

**Contents:**
- Schemas, indexes, statistics, security info, storage metadata.

**Role:**
- Guides parsing & optimization.
- Frequently cached for performance.

**Persistence:**
- Stored in the database but cached in memory.

---

## 8️⃣ Access Paths

### Heap Organization
- No ordering.
- Fast inserts.
- Full scans for non-key lookups.

### Clustered Organization
- Physical order by key.
- Great for range queries.
- Inserts may cause splits/moves.

### Unclustered Index
- Heap order + index mapping.
- Fast lookups but more random I/O.

**Trade-offs:**
- Reads → Clustered.
- Writes → Heap.

---

## 9️⃣ Integration of Concepts
Typical request flow:
1. **Query Processor** — SQL → physical plan.
2. **Execution Engine** — Requests pages from storage.
3. **Buffer Pool** — Returns from memory or loads from disk.
4. **Slotted Page** — Locates record within page.
5. **Catalog** — Validates schema & storage mapping.
6. **Access Path** — Chooses optimal data traversal.

## 1️⃣ DBMS Components Simulated in Code
The program is structured into **three main processes (classes)**, similar to a real DBMS:

### **`SlottedPage`** – *Represents a single physical page in storage*
- Holds multiple variable-length records.
- Uses a slot directory to track record positions.
- Keeps track of free space inside the page.

### **`BufferPool`** – *Simulates the memory manager of a DBMS*
- Stores recently used pages in memory.
- Uses **LRU (Least Recently Used)** replacement when memory is full.
- Writes **dirty pages** back to disk to preserve changes.

### **`Catalog`** – *The data dictionary / metadata manager*
- Stores table definitions (names, columns, clustered key).
- Tracks which pages belong to each table.
- Handles record insertion and retrieval.

---

## 2️⃣ Storage Models Implemented
The code supports **two types of access paths**:
- **Heap storage** – Records are stored in the order they arrive.
- **Clustered storage** – Records are sorted on insertion based on a **clustered key** (e.g., `name`).

---

## 3️⃣ How Data is Stored
When you insert a record:
1. **Catalog** converts the record dictionary into a binary format using `struct.pack()` for storage in pages.
2. **BufferPool** loads the correct page into memory (or creates a new one if needed).
3. **SlottedPage** inserts the record into free space and updates the slot directory.
4. If the page is **full**, a new page is allocated.
5. When the buffer pool runs out of memory slots, **LRU eviction** writes the least recently used page to disk.

---

## 4️⃣ Disk I/O Simulation
- Pages are saved to `.bin` files on disk using Python’s **`pickle`** module.
- When pages are loaded from disk, they are restored exactly as they were before.
- This models how a real DBMS swaps pages between **disk storage** and **main memory**.

---

## 5️⃣ Output Behavior
- **Heap table** → Rows are stored in **insert order** (unordered storage).
- **Clustered table** → Rows are **sorted** by the clustered key (e.g., `name`).

**Example Output:**
```sql
Heap Table Records:
Student0 → Student4 (in insert order)

Clustered Table Records:
Alex, Beatrice, John, Mohan, Zara (alphabetical order)


````python
"""
Week 1: Database Systems Architecture & Engine Foundations
----------------------------------------------------------
This project simulates the internal storage layer of a DBMS.
It includes:
- Slotted pages for storing variable-length records
- Buffer pool with LRU page replacement
- Disk persistence (pages stored as binary files)
- Catalog (data dictionary) for table metadata
- Heap and clustered access paths
"""

# Import Python libraries for required functionality
import os                  # For file and directory handling (disk storage simulation)
import struct              # For packing/unpacking binary data (record format)
import uuid                # For generating unique record IDs
import pickle              # For serializing pages to/from disk
from collections import OrderedDict  # For maintaining LRU order in buffer pool


# =========================================================
# SlottedPage Class - Represents a single physical storage page
# =========================================================
class SlottedPage:
    def __init__(self, page_id, page_size=4096):
        self.page_id = page_id                # Unique ID for this page (string)
        self.page_size = page_size            # Page size in bytes (default: 4KB)
        self.data = bytearray(page_size)      # Raw page data (initialized with zeros)
        self.slot_directory = []              # Stores (offset, length) for each record
        self.free_space_pointer = page_size   # Points to start of free space from the end of page
        self.dirty = False                    # Flag indicating if page has unsaved changes

    def insert_record(self, record_bytes):
        """
        Insert a serialized record into the page.
        - record_bytes: bytes representing a single record
        """
        record_length = len(record_bytes)  # Calculate length of record

        # Check if there is enough free space for record + slot entry
        if self.free_space_pointer - (len(self.slot_directory) + 1) * 4 < record_length:
            raise Exception("Not enough space in page")

        # Move free space pointer backward to allocate space
        self.free_space_pointer -= record_length

        # Store record data into the page's free space area
        self.data[self.free_space_pointer:self.free_space_pointer + record_length] = record_bytes

        # Add new slot entry containing offset and length
        self.slot_directory.append((self.free_space_pointer, record_length))

        # Mark page as modified
        self.dirty = True

    def get_record(self, slot_id):
        """
        Retrieve record bytes from a specific slot ID.
        """
        # Validate slot ID
        if slot_id < 0 or slot_id >= len(self.slot_directory):
            raise Exception("Invalid slot ID")

        # Retrieve offset and length from slot directory
        offset, length = self.slot_directory[slot_id]

        # Return the raw bytes of the record
        return self.data[offset:offset + length]

    def to_bytes(self):
        """
        Serialize the page object to bytes for writing to disk.
        """
        return pickle.dumps({
            "page_id": self.page_id,
            "page_size": self.page_size,
            "data": self.data,
            "slot_directory": self.slot_directory,
            "free_space_pointer": self.free_space_pointer
        })

    @staticmethod
    def from_bytes(byte_data):
        """
        Deserialize bytes from disk into a SlottedPage object.
        """
        obj = pickle.loads(byte_data)  # Load stored dictionary from bytes
        page = SlottedPage(obj["page_id"], obj["page_size"])  # Create new page instance
        page.data = obj["data"]  # Restore raw data
        page.slot_directory = obj["slot_directory"]  # Restore slot directory
        page.free_space_pointer = obj["free_space_pointer"]  # Restore free space pointer
        return page

    def __repr__(self):
        """
        String representation for debugging (shows slots, free space, dirty flag).
        """
        return f"<Page {self.page_id} slots={len(self.slot_directory)} free={self.free_space_pointer} dirty={self.dirty}>"


# =========================================================
# BufferPool Class - Manages in-memory pages and disk I/O
# =========================================================
class BufferPool:
    def __init__(self, capacity=3, storage_path="db_pages"):
        self.capacity = capacity               # Max number of pages to keep in memory
        self.pages = OrderedDict()              # Maintains pages in LRU order
        self.storage_path = storage_path        # Directory to store page files
        os.makedirs(storage_path, exist_ok=True)  # Create directory if it doesn't exist

    def get_page(self, page_id):
        """
        Retrieve a page by ID, loading from disk if not in memory.
        """
        # If page is already in memory, move it to the end (most recently used)
        if page_id in self.pages:
            self.pages.move_to_end(page_id)
            return self.pages[page_id]

        # If page exists on disk, load it
        page_path = os.path.join(self.storage_path, f"{page_id}.bin")
        if os.path.exists(page_path):
            with open(page_path, "rb") as f:
                page = SlottedPage.from_bytes(f.read())
        else:
            # Create a new blank page if it doesn't exist yet
            page = SlottedPage(page_id)

        # Add the page to buffer pool (evict if needed)
        self._add_to_pool(page)
        return page

    def _add_to_pool(self, page):
        """
        Add a page to the pool, evicting the least recently used (LRU) if full.
        """
        if len(self.pages) >= self.capacity:
            evict_id, evict_page = self.pages.popitem(last=False)  # Remove LRU page
            if evict_page.dirty:
                self._write_page_to_disk(evict_page)  # Save changes before eviction
        self.pages[page.page_id] = page  # Add new page to memory

    def _write_page_to_disk(self, page):
        """
        Save a page to disk as a binary file.
        """
        page_path = os.path.join(self.storage_path, f"{page.page_id}.bin")
        with open(page_path, "wb") as f:
            f.write(page.to_bytes())  # Serialize and write page
        page.dirty = False  # Mark as clean after writing

    def flush_all(self):
        """
        Write all dirty pages in the pool to disk.
        """
        for page in self.pages.values():
            if page.dirty:
                self._write_page_to_disk(page)


# =========================================================
# Catalog Class - Stores table metadata (data dictionary)
# =========================================================
class Catalog:
    def __init__(self, buffer_pool):
        self.buffer_pool = buffer_pool  # Reference to buffer pool for page management
        self.tables = {}                # Dictionary to store table metadata

    def create_table(self, table_name, columns, clustered_key=None):
        """
        Create a new table with specified columns and optional clustered key.
        """
        if table_name in self.tables:
            raise Exception("Table already exists")
        self.tables[table_name] = {
            "columns": columns,              # List of column names
            "pages": [f"{table_name}_0"],     # List of page IDs (start with one)
            "clustered_key": clustered_key    # Optional key for sorted storage
        }

    def insert_into_table(self, table_name, record_dict):
        """
        Insert a record into a table, handling heap or clustered storage.
        """
        if table_name not in self.tables:
            raise Exception("Table not found")

        table_meta = self.tables[table_name]
        record_bytes = self._serialize(record_dict, table_meta["columns"])  # Convert to bytes

        # If clustered storage is used
        if table_meta["clustered_key"]:
            page_id = table_meta["pages"][0]
            page = self.buffer_pool.get_page(page_id)
            # Get all records, add new one, and sort
            records = self.select_all(table_name)
            records.append(record_dict)
            records.sort(key=lambda r: r[table_meta["clustered_key"]])
            # Clear and reinsert sorted records
            page.slot_directory.clear()
            page.free_space_pointer = page.page_size
            for r in records:
                page.insert_record(self._serialize(r, table_meta["columns"]))
        else:
            # Heap storage: insert into first page with space
            inserted = False
            for page_id in table_meta["pages"]:
                page = self.buffer_pool.get_page(page_id)
                try:
                    page.insert_record(record_bytes)
                    inserted = True
                    break
                except:
                    continue
            # If no page had space, create a new one
            if not inserted:
                new_page_id = f"{table_name}_{len(table_meta['pages'])}"
                new_page = self.buffer_pool.get_page(new_page_id)
                new_page.insert_record(record_bytes)
                table_meta["pages"].append(new_page_id)

    def select_all(self, table_name):
        """
        Retrieve all records from a table.
        """
        if table_name not in self.tables:
            raise Exception("Table not found")

        table_meta = self.tables[table_name]
        results = []
        for page_id in table_meta["pages"]:
            page = self.buffer_pool.get_page(page_id)
            for slot_id in range(len(page.slot_directory)):
                raw = page.get_record(slot_id)
                results.append(self._deserialize(raw, table_meta["columns"]))
        return results

    def _serialize(self, record_dict, columns):
        """
        Convert a record dictionary into a binary format with length-prefix encoding.
        """
        record_bytes = b""
        for col in columns:
            val = str(record_dict[col]).encode("utf-8")     # Convert value to bytes
            record_bytes += struct.pack("H", len(val)) + val  # Store length + data
        return record_bytes

    def _deserialize(self, record_bytes, columns):
        """
        Convert binary data back into a record dictionary.
        """
        record = {}
        offset = 0
        for col in columns:
            length = struct.unpack("H", record_bytes[offset:offset+2])[0]  # Read length
            offset += 2
            val = record_bytes[offset:offset+length].decode("utf-8")       # Read value
            offset += length
            record[col] = val
        return record


# =========================================================
# DEMO: Using the Mini Storage Engine
# =========================================================
if __name__ == "__main__":
    # 1. Initialize buffer pool with max 3 pages in memory
    buffer_pool = BufferPool(capacity=3)

    # 2. Create catalog (data dictionary manager)
    catalog = Catalog(buffer_pool)

    # 3. Create two tables: one heap storage, one clustered by name
    catalog.create_table("students_heap", ["id", "name", "age"])
    catalog.create_table("students_clustered", ["id", "name", "age"], clustered_key="name")

    # 4. Insert into heap table (unordered)
    for i in range(5):
        catalog.insert_into_table("students_heap", {
            "id": str(uuid.uuid4())[:8],  # Short unique ID
            "name": f"Student{i}",
            "age": str(18 + i)
        })

    # 5. Insert into clustered table (sorted by name)
    names = ["Zara", "Alex", "Mohan", "Beatrice", "John"]
    for i, name in enumerate(names):
        catalog.insert_into_table("students_clustered", {
            "id": str(uuid.uuid4())[:8],
            "name": name,
            "age": str(18 + i)
        })

    # 6. Read and print all records from heap table
    print("\nHeap Table Records:")
    for r in catalog.select_all("students_heap"):
        print(r)

    # 7. Read and print all records from clustered table (sorted)
    print("\nClustered Table Records (sorted by name):")
    for r in catalog.select_all("students_clustered"):
        print(r)

    # 8. Write all dirty pages in buffer pool to disk
    buffer_pool.flush_all()
````

## Output Analysis

### Heap Table Records
- The five student entries are stored **exactly** in the order they were inserted.  
- No sorting is done , typical for **heap-based storage**, where pages simply accept the next available record in sequence.

### Clustered Table Records
- The same five students are **sorted alphabetically by name** before being stored.  
- This demonstrates the **clustered index** behavior, where the table’s physical order matches the order of the clustered key.  
- This enables **faster range queries** on that column.
