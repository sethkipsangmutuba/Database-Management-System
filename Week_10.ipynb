# Week 10: NoSQL & NewSQL Architectures

## 1. Introduction

Traditional relational databases (RDBMS) have long dominated enterprise applications due to their strong consistency guarantees and structured query language (SQL). However, with the rise of large-scale, distributed, and high-performance web applications, new data models and architectures have emerged — namely NoSQL and NewSQL systems. These database systems are designed to handle scalability, flexible data formats, and high throughput at the expense of certain RDBMS features.

This week explores NoSQL and NewSQL architectures, their design principles, trade-offs, and when to use each. We also examine key types of NoSQL systems—key-value stores, document databases, graph databases, and wide-column stores—as well as emerging NewSQL systems that aim to blend SQL consistency with NoSQL scalability.

## 2. Evolution Beyond Relational Databases

### 2.1 Why RDBMS Fall Short at Scale

RDBMSs enforce strong consistency and ACID properties, but struggle with:

- Horizontal scalability  
- Flexibility in handling semi-structured/unstructured data  
- High-velocity data ingestion  
- Distributed deployments across datacenters or clouds  

### 2.2 Emergence of NoSQL

NoSQL (Not Only SQL) systems emerged in the 2000s to overcome these limitations. NoSQL databases relax some RDBMS constraints (especially schema rigidity and consistency) to offer high scalability and availability.

### 2.3 Rise of NewSQL

NewSQL systems seek to provide the scalability and distribution of NoSQL systems while retaining SQL features and ACID guarantees.

## 3. NoSQL Database Categories

NoSQL systems are categorized by their data models and internal architectures. The four major types are:

### 3.1 Key-Value Stores

Key-value stores are the simplest type of NoSQL database. Data is stored as key-value pairs, like a large distributed hash table.

#### 3.1.1 Example: RocksDB

- Embedded key-value store developed by Facebook  
- Optimized for fast write-heavy workloads  
- Uses LSM-trees (Log-Structured Merge Trees) to organize and persist data efficiently  

#### 3.1.2 LSM-Trees and Compaction

- LSM-trees buffer writes in memory, periodically flushing to disk  
- Compaction merges data segments, removes duplicates, and maintains sorted order  
- Benefits include write optimization and compression  

**Use Cases**  
- Session management  
- Caching  
- Queues and real-time analytics  

### 3.2 Document Databases

Document databases store data in formats like JSON, BSON, or XML. Each document is a self-contained unit of data with flexible schema.

#### 3.2.1 Example: MongoDB

- Popular open-source document-oriented DBMS  
- Stores documents as BSON (binary JSON)  
- Supports indexing, replication, sharding  

#### 3.2.2 Indexing in MongoDB

- Indexes on fields inside documents improve query performance  
- Supports compound indexes, full-text search, and geospatial queries  

**Use Cases**  
- Content management systems  
- Real-time analytics  
- Catalogs and product inventories  

### 3.3 Graph Databases

Graph databases model data as nodes and relationships, making them ideal for highly connected data.

#### 3.3.1 Example: Neo4j

- A leading graph database  
- Uses a property graph model with nodes, edges, and properties  
- Supports Gremlin and Cypher query languages  

#### 3.3.2 Graph Querying with Gremlin

- Gremlin is a graph traversal language  
- Allows pattern matching, path finding, and analytics on complex networks  

**Use Cases**  
- Social networks  
- Fraud detection  
- Recommendation engines  

### 3.4 Wide-Column Stores

Wide-column stores store data in column families and support flexible schemas for each row.

#### 3.4.1 Example: Apache Cassandra

- Designed for high availability and scalability  
- Stores data across multiple nodes in a ring architecture  
- Supports tunable consistency, allowing trade-offs between consistency and latency  

#### 3.4.2 Internals of Cassandra

- Based on Amazon's Dynamo architecture and Bigtable model  
- Uses LSM-trees and compaction similar to RocksDB  
- Partitions data using consistent hashing  

**Use Cases**  
- Time-series data  
- Messaging systems  
- Internet of Things (IoT) platforms  

## 4. NewSQL Systems

NewSQL aims to provide the best of both worlds: SQL interface + ACID compliance + horizontal scalability.

### 4.1 Characteristics of NewSQL

- Fully support SQL (DDL, DML, joins, transactions)  
- Maintain strong consistency using distributed consensus  
- Scale-out on commodity hardware  
- Often built from scratch (not extensions of legacy RDBMS)  

### 4.2 Examples of NewSQL Systems

#### 4.2.1 CockroachDB

- Inspired by Google Spanner  
- Uses Raft consensus protocol for strong consistency  
- Globally distributed and strongly consistent  
- PostgreSQL-compatible interface  

#### 4.2.2 FaunaDB

- Serverless and globally consistent  
- Combines document and relational features  
- Uses Calvin protocol for transaction coordination  
- Suitable for modern, cloud-native applications  

## 5. Comparing SQL, NoSQL, and NewSQL

| Feature | RDBMS (SQL) | NoSQL | NewSQL |
|--------|-------------|--------|---------|
| Schema | Fixed (rigid) | Flexible or schema-less | Flexible or semi-structured |
| Consistency | Strong (ACID) | Eventual / tunable | Strong (ACID) |
| Scalability | Vertical | Horizontal | Horizontal |
| Query Language | SQL | Custom (e.g., Gremlin) | SQL |
| Transaction Support | Full (ACID) | Limited / BASE | Full (ACID) |
| Use Cases | Financial, ERP | Web apps, IoT, Big Data | Cloud apps, modern SaaS |

### 5.1 Trade-Offs

- **NoSQL**: Great for flexible schemas, horizontal scaling, and high-velocity workloads but lacks strong consistency and SQL expressiveness.  
- **NewSQL**: Ideal for distributed apps needing transactional guarantees, but may have a steeper learning curve or performance trade-offs in some cases.  
- **RDBMS**: Stable and mature for traditional enterprise systems but limited in distributed and high-volume environments.  

## 6. Use Case Scenarios

```python
# === Week 10: NoSQL & NewSQL Architectures Simulation ===

class KeyValueStore:
    def __init__(self):
        self.memtable = {}
        self.sstable = {}

    def put(self, key, value):
        self.memtable[key] = value
        if len(self.memtable) >= 5:
            self.flush_to_disk()

    def get(self, key):
        return self.memtable.get(key) or self.sstable.get(key)

    def flush_to_disk(self):
        print("Flushing memtable to disk...")
        self.sstable.update(self.memtable)
        self.memtable.clear()


class MongoDBSimulator:
    def __init__(self):
        self.documents = []
        self.index = {}

    def insert(self, document):
        self.documents.append(document)
        for key, val in document.items():
            self.index.setdefault((key, val), []).append(document)

    def query(self, criteria):
        results = []
        for doc in self.documents:
            if all(doc.get(k) == v for k, v in criteria.items()):
                results.append(doc)
        return results


class Graph:
    def __init__(self):
        self.edges = {}

    def add_edge(self, src, dst):
        self.edges.setdefault(src, []).append(dst)

    def gremlin_traverse(self, start_node, depth=1):
        visited = set()
        frontier = [start_node]
        for _ in range(depth):
            next_frontier = []
            for node in frontier:
                visited.add(node)
                for neighbor in self.edges.get(node, []):
                    if neighbor not in visited:
                        next_frontier.append(neighbor)
            frontier = next_frontier
        return visited


class WideColumnStore:
    def __init__(self):
        self.data = {}

    def insert(self, row_key, column_family, column, value):
        self.data.setdefault(row_key, {}).setdefault(column_family, {})[column] = value

    def get(self, row_key, column_family, column):
        return self.data.get(row_key, {}).get(column_family, {}).get(column)


class NewSQLSimulator:
    def __init__(self):
        self.table = []

    def insert(self, row):
        self.table.append(row)

    def select(self, condition_fn):
        return [row for row in self.table if condition_fn(row)]


def simulate_key_value_store():
    print("=== Simulating Key-Value Store (RocksDB) ===")
    kv = KeyValueStore()
    for i in range(10):
        kv.put(f"session{i}", f"user_{i}")
    print("Get session5:", kv.get("session5"))


def simulate_mongodb_indexing():
    print("\n--- Lab: MongoDB Indexing ---")
    mongo = MongoDBSimulator()
    mongo.insert({"name": "Alice", "age": 30})
    mongo.insert({"name": "Bob", "age": 25})
    mongo.insert({"name": "Carol", "age": 30})
    result = mongo.query({"age": 30})
    print("Documents with age 30:", result)


def simulate_gremlin_graph():
    print("\n--- Lab: Gremlin Traversal on Graph ---")
    g = Graph()
    g.add_edge("Alice", "Bob")
    g.add_edge("Bob", "Carol")
    g.add_edge("Carol", "Dave")
    traversed = g.gremlin_traverse("Alice", depth=2)
    print("Nodes reachable from Alice in 2 steps:", traversed)


def simulate_wide_column_store():
    print("\n--- Simulating Wide Column Store (Cassandra) ---")
    cass = WideColumnStore()
    cass.insert("user123", "profile", "name", "Alice")
    cass.insert("user123", "profile", "email", "alice@example.com")
    name = cass.get("user123", "profile", "name")
    print("User name:", name)


def simulate_newsql_system():
    print("\n--- Simulating NewSQL Query (CockroachDB/FaunaDB) ---")
    db = NewSQLSimulator()
    db.insert({"id": 1, "product": "Book", "price": 15})
    db.insert({"id": 2, "product": "Pen", "price": 5})
    result = db.select(lambda row: row["price"] > 10)
    print("Products with price > 10:", result)


def run_all_simulations():
    print("===== Week 10: NoSQL & NewSQL Architectures Simulation =====")
    simulate_key_value_store()
    simulate_mongodb_indexing()
    simulate_gremlin_graph()
    simulate_wide_column_store()
    simulate_newsql_system()


if __name__ == "__main__":
    run_all_simulations()
```
