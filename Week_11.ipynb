# Week 11: Streaming Systems & Real-Time Processing

## 1. Introduction  
In an increasingly data-driven world, the ability to process information as it is generated has become critical. Traditional batch processing systems are insufficient for scenarios that require immediate response — such as fraud detection, stock trading, traffic monitoring, or social media analytics. This has led to the emergence of streaming systems, which are capable of ingesting, processing, and analyzing data in real time or near real time.  

This week's lecture explores the fundamental concepts of real-time data processing, key components of modern streaming architectures, and compares leading frameworks such as Apache Kafka, Apache Flink, and Apache Spark Streaming. Core concepts such as windowing, event-time processing, watermarking, and handling late data are also examined. Finally, the debate between Kappa and Lambda architectures is introduced.

## 2. The Need for Real-Time Processing

### 2.1 Limitations of Batch Processing  
Batch systems (e.g., Hadoop MapReduce) are efficient for large-scale data processing but have high latency. They are unsuitable for scenarios where insights or actions must be taken within seconds or milliseconds.  

**Drawbacks:**  
- Delayed response time  
- Poor handling of continuous or unbounded data streams  
- Complex retry/reprocessing logic for real-time use cases  

### 2.2 Use Cases Requiring Streaming  
- Financial fraud detection  
- Sensor data monitoring (IoT)  
- Social media trend analysis  
- Real-time marketing and personalization  
- Online gaming telemetry  
- Traffic and weather forecasting  

## 3. Streaming Data Models  
A data stream is a sequence of unbounded, continuous data events. Unlike batches, streaming data is often:  
- Time-sensitive  
- Unpredictable in volume  
- Unstructured or semi-structured  

### 3.1 Key Concepts in Stream Processing  
| Concept        | Description                                         |
|----------------|-----------------------------------------------------|
| Event          | A single data record (e.g., a log, sensor reading) |
| Stream         | A continuous flow of events over time              |
| Time Semantics | Differentiates between event time, processing time, and ingestion time |
| Windowing      | Technique to segment unbounded streams into finite datasets |
| Watermarks     | Signals to the system about progress in event time, useful for late data |

## 4. Time in Stream Processing  
Handling time correctly is critical for accurate analytics in streaming systems.

### 4.1 Event Time vs Processing Time  
| Type            | Definition                                        |
|------------------|--------------------------------------------------|
| Event Time       | The actual time the event occurred (e.g., on the device) |
| Processing Time  | The time when the event is processed by the system |
| Ingestion Time   | The time when the system receives the event     |

Event time is preferred in analytics since it reflects the true sequence of events, but it introduces complexity such as out-of-order events.

## 5. Windowing in Streaming Systems  
Windowing is a method for grouping events over a period to make stream processing computationally feasible and meaningful.

### 5.1 Types of Windows  
| Window Type      | Description                                       |
|------------------|---------------------------------------------------|
| Tumbling Windows | Fixed, non-overlapping windows (e.g., every 5 minutes) |
| Sliding Windows  | Overlapping windows sliding over time (e.g., every 5 minutes with a 1-minute slide) |
| Session Windows  | Variable-sized windows based on user activity sessions |
| Global Windows   | No window boundaries; used with triggers or custom logic |

## 6. Handling Late and Out-of-Order Data  
Real-time systems must manage events that arrive late due to network delays or clock skews.

### 6.1 Watermarking  
Watermarks are timestamps that inform the system about the progress of event time. They help:  
- Determine when to close a window  
- Handle out-of-order data  
- Trigger calculations and aggregations  

### 6.2 Late Data Policies  
- Drop late data entirely  
- Redirect to a side output for later processing  
- Update previous results (if system supports stateful computation)  

## 7. Key Streaming Technologies  

### 7.1 Apache Kafka  
Kafka is a distributed streaming platform that serves as the backbone of many real-time architectures.  

**Key features:**  
- Durable message storage  
- High throughput and fault-tolerance  
- Pub/sub and queue messaging models  
- Integration with stream processors like Flink and Spark  

**Kafka Roles:**  
- **Producer:** Sends events to Kafka  
- **Broker:** Stores streams (topics) in partitions  
- **Consumer:** Reads and processes events from topics  

### 7.2 Apache Flink  
Flink is a powerful stream processing engine designed for stateful computations over unbounded data.  

**Core features:**  
- Event-time processing and watermarks  
- Complex event processing (CEP)  
- Exactly-once semantics  
- Low latency and high throughput  
- Supports both stream and batch jobs (unified runtime)  

Flink is ideal for advanced analytics, stateful aggregations, and handling out-of-order events.

### 7.3 Apache Spark Streaming  
Spark Streaming extends the Spark engine for micro-batch stream processing.  

**Key aspects:**  
- Treats stream data as small batches  
- Uses DStreams (discretized streams)  
- Supports high-level APIs for map, reduce, join, etc.  
- Integrates with Spark SQL and MLlib  

While less responsive than Flink in ultra-low-latency settings, Spark Streaming benefits from the broader Spark ecosystem and ease of use.

## 8. Stream Processing Semantics  
| Semantics       | Meaning                                          |
|------------------|--------------------------------------------------|
| At-most-once     | Each event processed zero or one time (may miss) |
| At-least-once    | Each event processed one or more times (may duplicate) |
| Exactly-once     | Each event processed exactly one time (ideal, costly) |

Trade-offs depend on use case. For billing or financial systems, exactly-once is critical; for social media feeds, at-least-once is often sufficient.

## 9. Kappa vs Lambda Architectures  

### 9.1 Lambda Architecture  
Introduced by Nathan Marz, Lambda handles data through two paths:  
- **Batch Layer:** Processes large historical data using batch systems (e.g., Hadoop)  
- **Speed Layer:** Processes real-time data using streaming engines (e.g., Spark Streaming)  
- **Serving Layer:** Merges results from both layers  

**Benefits:** Accuracy from batch; freshness from stream  
**Challenges:**  
- Code duplication across layers  
- Operational complexity  

### 9.2 Kappa Architecture  
Proposed by Jay Kreps (creator of Kafka), Kappa simplifies processing:  
- Single streaming engine handles all data  
- Reprocessing is done by replaying logs from Kafka  

**Benefits:**  
- Simpler architecture  
- Easier maintenance  

**Limitations:**  
- Full reprocessing can be expensive  
- Streaming engine must be highly capable (like Flink)  

## 10. Challenges in Real-Time Stream Processing  

### 10.1 Scalability  
- Requires horizontal scalability to handle millions of events per second  
- Partitioning and parallel processing are key  

### 10.2 Fault Tolerance  
- Must recover from node crashes, data loss, or network failures  
- Checkpointing and replication are common solutions  

### 10.3 State Management  
- Stateful operators maintain memory of past events  
- Needs mechanisms for persisting and restoring state across failures  

### 10.4 Data Quality  
- Ensuring accuracy and completeness is hard with late or out-of-order data

```python
from datetime import datetime, timedelta
import random
import time

# Simulated Event Stream
class Event:
    def __init__(self, user_id, action, event_time):
        self.user_id = user_id
        self.action = action
        self.event_time = event_time

    def __repr__(self):
        return f"Event(user_id={self.user_id}, action={self.action}, event_time={self.event_time.strftime('%H:%M:%S')})"


class EventStreamSimulator:
    def __init__(self):
        self.actions = ["click", "view", "purchase", "scroll"]
        self.current_time = datetime.now()

    def generate_event(self):
        self.current_time += timedelta(seconds=1)
        return Event(
            user_id=random.randint(1, 5),
            action=random.choice(self.actions),
            event_time=self.current_time
        )


# Windowing System
class TumblingWindow:
    def __init__(self, window_size_seconds):
        self.window_size = timedelta(seconds=window_size_seconds)
        self.current_window_start = None
        self.events = []

    def add_event(self, event):
        if self.current_window_start is None:
            self.current_window_start = event.event_time

        if event.event_time >= self.current_window_start + self.window_size:
            self.process_window()
            self.current_window_start = event.event_time
            self.events = []

        self.events.append(event)

    def process_window(self):
        print(f"\nProcessing window starting at {self.current_window_start.strftime('%H:%M:%S')}")
        counts = {}
        for event in self.events:
            counts[event.action] = counts.get(event.action, 0) + 1
        for action, count in counts.items():
            print(f"  {action}: {count} events")


# Watermarking and Late Data
class WatermarkingSystem:
    def __init__(self, allowed_lateness_seconds):
        self.allowed_lateness = timedelta(seconds=allowed_lateness_seconds)
        self.latest_event_time = None

    def update_watermark(self, event_time):
        if self.latest_event_time is None or event_time > self.latest_event_time:
            self.latest_event_time = event_time

    def is_late(self, event_time):
        if self.latest_event_time is None:
            return False
        return event_time < self.latest_event_time - self.allowed_lateness


# Real-time Simulation
def simulate_real_time_streaming():
    print("=== Week 11: Streaming Systems & Real-Time Processing Simulation ===\n")
    stream = EventStreamSimulator()
    window = TumblingWindow(window_size_seconds=5)
    watermarking = WatermarkingSystem(allowed_lateness_seconds=2)

    for _ in range(20):
        event = stream.generate_event()

        # Simulate random delay (simulate out-of-order events)
        if random.random() < 0.2:
            delay = timedelta(seconds=random.randint(1, 3))
            event.event_time -= delay
            print(f"Late Event Simulated (Delayed by {delay.seconds}s): {event}")

        watermarking.update_watermark(event.event_time)
        if watermarking.is_late(event.event_time):
            print(f"  --> Dropped Late Event: {event}")
        else:
            window.add_event(event)

        time.sleep(0.1)  # simulate time passing

    window.process_window()  # process any remaining events


# Run Simulation
if __name__ == "__main__":
    simulate_real_time_streaming()
```
