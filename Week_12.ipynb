# Week 12: Security, Privacy & Compliance in Databases

## 1. Introduction  
Data is a vital asset in today’s digital economy, and with the proliferation of sensitive personal and organizational data, ensuring the security, privacy, and legal compliance of database systems is more important than ever. Database security aims to protect data from unauthorized access, misuse, or corruption. Privacy focuses on how data about individuals is collected, stored, and shared. Compliance relates to adhering to laws and standards governing the handling of data.  

In this week, we study role-based access control (RBAC), row-level security, transparent data encryption (TDE), data masking, differential privacy, $k$-anonymity, and key data protection laws like GDPR, HIPAA, and CCPA. Understanding these elements allows database professionals to build systems that are not only functional but also safe and legally compliant.

## 2. Database Security Fundamentals

### 2.1 The CIA Triad  
Security in databases is often structured around the CIA triad:

- **Confidentiality** – ensuring data is accessible only to authorized users.  
- **Integrity** – maintaining accuracy and consistency of data.  
- **Availability** – ensuring reliable access to data when needed.

Each component must be balanced in database system design.

## 3. Access Control Mechanisms

### 3.1 Role-Based Access Control (RBAC)  
RBAC is a security model where permissions are not assigned directly to users, but to roles, which are then assigned to users.

**Components of RBAC:**

- **Users** – Individuals who use the system.  
- **Roles** – Named collections of permissions (e.g., admin, analyst).  
- **Permissions** – Operations allowed (e.g., SELECT, INSERT).  
- **Sessions** – Mapping between a user and activated roles.

**Benefits:**

- Simplifies management of user privileges.  
- Enforces the principle of least privilege.  
- Scales better in large organizations.

**Example Use Case:** A healthcare database might have roles such as Doctor, Nurse, and Receptionist, each with specific data access privileges.

### 3.2 Row-Level Security (RLS)  
RLS is a finer-grained access control mechanism that restricts access to specific rows within a table, based on user identity or role.

**How RLS works:**

- Policies are defined per table.  
- Policies filter rows automatically based on current user/session.

**Use Cases:**

- Multitenant systems (each user sees only their own data).  
- Sensitive data segregation (e.g., doctors see only their patients’ records).

**Advantages:**

- Enhanced data isolation.  
- Reduces reliance on application-side filters (which can be bypassed).

## 4. Data Protection Techniques

### 4.1 Transparent Data Encryption (TDE)  
TDE encrypts the entire database or tablespaces at the storage level, making data unreadable to unauthorized users, even if the disk is stolen.

**Key Features:**

- Transparent to applications.  
- Protects data at rest.  
- Requires secure key management.

**Limitations:**

- Does not protect against insider threats or runtime access.  
- May introduce slight performance overhead.

**Vendors Supporting TDE:** Oracle, Microsoft SQL Server, PostgreSQL (via extensions), MySQL.

### 4.2 Data Masking  
Data masking replaces sensitive data with obfuscated, yet realistic values to protect it in non-production environments.

**Types of Masking:**

- **Static masking** – masked data is stored and used for testing or analytics.  
- **Dynamic masking** – data is masked on-the-fly during queries.

**Use Case:** Masking customer SSNs before sharing databases with development teams.

**Benefits:**

- Protects sensitive data during testing or training.  
- Helps meet compliance without exposing real data.

## 5. Privacy-Preserving Techniques

### 5.1 Differential Privacy  
Differential privacy ensures that statistical queries on datasets do not compromise the privacy of any individual record.

**Core Principle:** Adding calibrated noise to results so that the inclusion or exclusion of a single data point cannot be detected.

**Mathematical Property:**  
An algorithm is $\varepsilon$-differentially private if the probability of any output does not differ significantly with or without any single individual's data.

**Advantages:**

- Strong theoretical guarantees.  
- Increasing adoption by tech companies and governments.

**Challenges:**

- Choosing the right level of noise (trade-off between privacy and utility).  
- Complexity of implementation.

### 5.2 K-Anonymity  
$k$-anonymity is a property where a dataset is transformed so that every individual is indistinguishable from at least $k - 1$ others based on identifying attributes.

**Key Concepts:**

- **Quasi-identifiers** – attributes that can indirectly identify a person (e.g., age, ZIP code).  
- **Generalization and suppression** – techniques used to achieve $k$-anonymity.

**Example:** A dataset with $k = 3$ ensures that each combination of quasi-identifiers appears at least 3 times.

**Limitations:**

- Vulnerable to homogeneity and background knowledge attacks.  
- May reduce data utility.

**Variants:**

- **$\ell$-diversity** – adds intra-group diversity to protect against inference.  
- **$t$-closeness** – ensures distribution of sensitive values is close to the whole dataset.

## 6. Legal and Regulatory Compliance  
Organizations are subject to various laws and standards based on the nature of data and jurisdictions in which they operate.

### 6.1 General Data Protection Regulation (GDPR)  
**Jurisdiction:** European Union

**Key Provisions:**

- Right to be forgotten  
- Data portability  
- Explicit consent for data processing  
- Fines up to €20 million or 4% of annual turnover

**Database Impact:**

- Must track and log data access  
- Enable data erasure  
- Support pseudonymization and encryption

### 6.2 Health Insurance Portability and Accountability Act (HIPAA)  
**Jurisdiction:** United States (healthcare)

**Main Goals:**

- Protect patient health information (PHI)  
- Establish security and privacy rules

**Database Requirements:**

- Access controls and audit logging  
- Data encryption  
- Breach notification procedures

### 6.3 California Consumer Privacy Act (CCPA)  
**Jurisdiction:** California, USA

**Provisions:**

- Right to know what data is collected  
- Right to delete personal information  
- Right to opt-out of data sale

**Compliance Strategies:**

- Maintain audit trails  
- Provide user data access portals  
- Implement opt-out mechanisms

## 7. Auditing and Monitoring

### 7.1 Database Auditing  
Auditing records who accessed what data and when. It helps in:

- Tracing unauthorized activity  
- Ensuring regulatory compliance  
- Supporting forensic investigations

**Types of Events Logged:**

- Login/logout attempts  
- Data read/write actions  
- Role or privilege changes  
- DDL changes

**Best Practices:**

- Log both success and failure events  
- Protect audit logs from tampering  
- Regularly review logs for anomalies

## 8. Best Practices for Secure and Compliant Databases

| Best Practice                      | Description                                      |
|-----------------------------------|--------------------------------------------------|
| Principle of Least Privilege      | Users should have the minimum access necessary   |
| Strong Authentication Mechanisms  | Use multi-factor authentication and password policies |
| Regular Patching                  | Apply security updates to fix vulnerabilities    |
| Encryption                        | Use TDE and SSL for data at rest and in transit  |
| Data Minimization                 | Store only the data you need, for the time needed |
| Secure Backups                    | Encrypt and protect backup files                 |
| Compliance Mapping                | Align database configurations with legal requirements |
| Periodic Audits and Penetration Testing | Identify vulnerabilities before they are exploited |

## 9. Emerging Trends in Database Security and Privacy

- **Zero Trust Security Models:** Assume breach and verify every access request.  
- **Confidential Computing:** Encrypt data during computation using secure enclaves.  
- **Homomorphic Encryption:** Enables computation on encrypted data without decryption.  
- **Synthetic Data Generation:** Replace real data with statistically similar fake data for training/testing.

```python
import base64
import random
import hashlib
from collections import defaultdict
import numpy as np

# ---------------------------
# ROLE-BASED ACCESS CONTROL (RBAC)
# ---------------------------
class RBACSystem:
    def __init__(self):
        self.roles = defaultdict(set)
        self.user_roles = defaultdict(set)

    def assign_role(self, user, role):
        self.user_roles[user].add(role)

    def grant_permission(self, role, permission):
        self.roles[role].add(permission)

    def check_access(self, user, permission):
        for role in self.user_roles[user]:
            if permission in self.roles[role]:
                return True
        return False

# ---------------------------
# ROW-LEVEL SECURITY (RLS)
# ---------------------------
class RowLevelSecurityDB:
    def __init__(self):
        self.data = []

    def add_row(self, user_id, row_data):
        self.data.append((user_id, row_data))

    def get_rows_for_user(self, user_id):
        return [row for uid, row in self.data if uid == user_id]

# ---------------------------
# TRANSPARENT DATA ENCRYPTION (TDE)
# ---------------------------
class TransparentEncryption:
    def __init__(self, key='secret'):
        self.key = key

    def encrypt(self, data):
        encoded = base64.b64encode(data.encode()).decode()
        return encoded

    def decrypt(self, encrypted_data):
        decoded = base64.b64decode(encrypted_data.encode()).decode()
        return decoded

# ---------------------------
# DATA MASKING
# ---------------------------
def mask_data(value, method="default"):
    if method == "default":
        return "****"
    elif method == "hash":
        return hashlib.sha256(value.encode()).hexdigest()[:8]
    elif method == "partial":
        return value[0] + "***" + value[-1]
    return value

# ---------------------------
# DIFFERENTIAL PRIVACY
# ---------------------------
def laplace_mechanism(true_value, epsilon=1.0):
    noise = np.random.laplace(0, 1/epsilon)
    return true_value + noise

# ---------------------------
# K-ANONYMITY
# ---------------------------
def k_anonymize(data, k=3):
    freq = defaultdict(int)
    for row in data:
        qid = (row['age']//10, row['zip'][:3])
        freq[qid] += 1

    anonymized = []
    for row in data:
        qid = (row['age']//10, row['zip'][:3])
        if freq[qid] >= k:
            row['age'] = f"{row['age']//10}0s"
            row['zip'] = row['zip'][:3] + "**"
        else:
            row['age'] = "**"
            row['zip'] = "****"
        anonymized.append(row)
    return anonymized

# ---------------------------
# MAIN SIMULATION
# ---------------------------
if __name__ == "__main__":
    print("=== Week 12: Security, Privacy & Compliance Simulation ===\n")

    # --- RBAC Simulation ---
    print("--- RBAC Example ---")
    rbac = RBACSystem()
    rbac.assign_role("alice", "doctor")
    rbac.grant_permission("doctor", "read_patient_data")

    print("Can Alice read patient data?", rbac.check_access("alice", "read_patient_data"))
    print("Can Alice delete patient data?", rbac.check_access("alice", "delete_patient_data"))

    # --- RLS Simulation ---
    print("\n--- Row-Level Security ---")
    db = RowLevelSecurityDB()
    db.add_row("alice", {"patient": "John", "diagnosis": "Flu"})
    db.add_row("bob", {"patient": "Jane", "diagnosis": "COVID-19"})
    print("Data visible to Alice:", db.get_rows_for_user("alice"))

    # --- Transparent Encryption ---
    print("\n--- Transparent Data Encryption (simulated) ---")
    tde = TransparentEncryption()
    secret_data = "Sensitive DB Info"
    encrypted = tde.encrypt(secret_data)
    decrypted = tde.decrypt(encrypted)
    print("Encrypted:", encrypted)
    print("Decrypted:", decrypted)

    # --- Data Masking ---
    print("\n--- Data Masking ---")
    print("Original SSN: 123-45-6789")
    print("Masked:", mask_data("123-45-6789", method="partial"))

    # --- Differential Privacy ---
    print("\n--- Differential Privacy ---")
    true_count = 50
    private_result = laplace_mechanism(true_count, epsilon=0.5)
    print(f"True count: {true_count}, DP Result: {round(private_result, 2)}")

    # --- K-Anonymity ---
    print("\n--- K-Anonymity ---")
    dataset = [
        {"name": "Alice", "age": 29, "zip": "12345"},
        {"name": "Bob", "age": 31, "zip": "12367"},
        {"name": "Carol", "age": 35, "zip": "12389"},
        {"name": "Dave", "age": 42, "zip": "98765"},
        {"name": "Eve", "age": 46, "zip": "98764"},
        {"name": "Frank", "age": 48, "zip": "98763"},
    ]
    anonymized_data = k_anonymize(dataset, k=3)
    for row in anonymized_data:
        print(row)
```
