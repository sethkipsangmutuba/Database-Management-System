### 2. Vector Databases

#### 2.1 What is a Vector Database?
A vector database is a specialized data management system designed to store, index, and search high-dimensional vector embeddings.

Embeddings are numerical representations of complex data — such as text, images, video, and audio — transformed by deep learning models into vectors in a high-dimensional space. These vectors allow similarity-based retrieval using nearest neighbor search techniques.

#### 2.2 Use Cases
- Semantic search in NLP: Searching documents by meaning rather than keywords  
- Image similarity: Finding visually similar images from a database  
- Recommendation systems: Matching users with similar preferences  
- Fraud detection: Identifying similar behavioral patterns  

#### 2.3 Examples of Vector Databases
- FAISS (Facebook AI Similarity Search)  
- Pinecone  
- Weaviate  
- Milvus  
- Qdrant  

#### 2.4 Approximate Nearest Neighbor (ANN) Search
In high dimensions, exact nearest neighbor search is computationally expensive. ANN techniques provide faster, scalable alternatives with minor accuracy trade-offs. Common algorithms:

- Hierarchical Navigable Small World (HNSW)  
- Product Quantization (PQ)  
- IVF (Inverted File Index)

### 3. Learned Indexes

#### 3.1 Motivation
Traditional index structures like B-Trees, Hash Maps, and Bloom Filters were designed under assumptions that do not always hold in modern data-intensive environments.

Learned Index Structures apply machine learning models to predict the position of data in storage, potentially outperforming handcrafted structures in:

- Space efficiency  
- Lookup latency  
- Cache friendliness  

#### 3.2 Concept of Learned Indexes
A learned index treats indexing as a regression problem: given a key, predict the position of the key in a sorted array.

If keys are evenly distributed, a simple linear model (like $y = mx + c$) can approximate the position. For complex distributions, hierarchical models are used.

#### 3.3 Recursive Model Index (RMI)
The Recursive Model Index (RMI) architecture includes:

- A root model that selects the correct second-level model  
- Leaf models that approximate positions for specific segments  
- Bounded error with fallback search (e.g., binary search)  

#### 3.4 Benefits and Challenges
**Advantages:**
- Smaller memory footprint  
- Faster lookups in predictable key distributions  
- Lower I/O overhead  

**Challenges:**
- Poor generalization to skewed data  
- Expensive retraining for dynamic workloads  
- Integration into existing DBMS architectures is non-trivial  

**Example Systems:**
- ALEX (Adaptive Learned Index)  
- PGM Index (Piecewise Geometric Models)

### 4. Multimodal Search

#### 4.1 Definition
Multimodal search enables querying across multiple data types — text, images, audio, and structured fields — simultaneously.

Example: Searching for "a red dress worn by a celebrity at an award show" using both image and text modalities.

#### 4.2 How It Works
- Embed each modality into a common vector space  
- Store embeddings in a vector database  
- Generate query vectors and match using similarity search  

**Techniques:**
- Contrastive learning (e.g., CLIP model)  
- Joint embedding models  

#### 4.3 Applications
- E-commerce product discovery  
- Medical diagnosis using image + patient records  
- Legal and academic search across documents and images

### 5. AI-Based Query Optimization

#### 5.1 Query Optimization in DBMS
Traditional optimizers use cost-based heuristics. Challenges include:

- Complex joins  
- Skewed distributions  
- Inaccurate cardinality estimates  
- Shifting data patterns  

#### 5.2 AI in Query Optimization
AI-enhanced optimizers use ML to improve plan selection.

**Techniques:**
- Reinforcement Learning (RL)  
- Supervised Learning  
- Graph Neural Networks (GNNs)  
- Bayesian Optimization  

#### 5.3 Examples of AI-Based Optimizers
- SkinnerDB – RL-based join ordering  
- Neo – Deep learning for join order prediction  
- Self-driving DBMS – Adaptive, learning query plans over time

#### 6.4 Explainability and Trust

```python
# Week 13: Frontier Topics Simulation — Vector DBs, Learned Indexes, AI

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
import random

# --- Vector Database Simulation (Semantic Search, ANN) ---
class VectorDatabase:
    def __init__(self):
        self.embeddings = {}

    def add(self, item_id, vector):
        self.embeddings[item_id] = np.array(vector)

    def search(self, query_vector, top_k=3):
        results = []
        for item_id, vec in self.embeddings.items():
            sim = np.dot(vec, query_vector) / (np.linalg.norm(vec) * np.linalg.norm(query_vector))
            results.append((item_id, sim))
        results.sort(key=lambda x: x[1], reverse=True)
        return results[:top_k]

# --- Learned Index Simulation ---
class LearnedIndex:
    def __init__(self, keys):
        self.sorted_keys = sorted(keys)
        X = np.arange(len(keys)).reshape(-1, 1)
        y = np.array(self.sorted_keys).reshape(-1, 1)
        self.model = LinearRegression().fit(y, X)

    def predict_position(self, key):
        pos = int(self.model.predict([[key]])[0][0])  # fixed deprecated warning
        pos = min(max(pos, 0), len(self.sorted_keys) - 1)
        return pos

    def lookup(self, key):
        pos = self.predict_position(key)
        for i in range(max(0, pos - 5), min(len(self.sorted_keys), pos + 5)):
            if self.sorted_keys[i] == key:
                return i
        return -1

# --- Multimodal Search Simulation ---
class MultimodalSearch:
    def __init__(self):
        self.items = {
            "Red dress - Oscars": np.random.rand(128),
            "Blue jeans - Mall": np.random.rand(128),
            "White suit - Gala": np.random.rand(128)
        }

    def search(self, query_vector):
        results = []
        for desc, vec in self.items.items():
            sim = np.dot(vec, query_vector) / (np.linalg.norm(vec) * np.linalg.norm(query_vector))
            results.append((desc, sim))
        results.sort(key=lambda x: x[1], reverse=True)
        return results[0]

# --- AI-Based Query Optimizer Simulation ---
class AIQueryOptimizer:
    def __init__(self):
        self.model = RandomForestClassifier()
        self.train_model()

    def train_model(self):
        X = [[random.randint(1, 100), random.randint(1, 100)] for _ in range(50)]
        y = [random.choice(["Plan_1", "Plan_2", "Plan_3"]) for _ in X]
        self.model.fit(X, y)

    def predict_best_plan(self, query_features):
        return self.model.predict([query_features])[0]

# --- Run All Simulations ---
def run_all_simulations():
    print("Week 13: Frontier Topics Simulation ===\n")

    # Vector DB
    print("--- Vector Database ---")
    vec_db = VectorDatabase()
    vec_db.add("img1", np.random.rand(128))
    vec_db.add("img2", np.random.rand(128))
    vec_db.add("img3", np.random.rand(128))
    query = np.random.rand(128)
    matches = vec_db.search(query)
    print("Top matches:", matches)
    print()

    # Learned Index
    print("--- Learned Index ---")
    index = LearnedIndex([10, 20, 30, 40, 50])
    print("Position of 40:", index.lookup(40))
    print("Position of 25:", index.lookup(25))
    print()

    # Multimodal Search
    print("--- Multimodal Search ---")
    mm_search = MultimodalSearch()
    result = mm_search.search(np.random.rand(128))
    print("Top match:", result)
    print()

    # AI-Based Query Optimization
    print("--- AI-Based Query Optimization ---")
    optimizer = AIQueryOptimizer()
    best_plan = optimizer.predict_best_plan([34, 80])
    print("Predicted best plan:", best_plan)
    print()

# Execute all
if __name__ == "__main__":
    run_all_simulations()
```
