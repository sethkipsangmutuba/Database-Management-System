# **Week 7: Object-Oriented, Deductive & XML Databases**

## **1. Introduction to Non-Relational Database Models**

While relational databases have dominated data management for decades, they are not always the most natural fit for all applications. Complex data representations, hierarchical relationships, logical inference, and semi-structured data often challenge the tabular model of relational databases.

This week’s focus explores three alternative paradigms:

- Object-Oriented Databases (OODBMS)  
- Deductive Databases (based on logic programming)  
- XML Databases (semi-structured data)  

Each of these models offers unique strengths in particular contexts such as software engineering, artificial intelligence, and web-based information systems.

---

## **2. Object-Oriented Databases (OODBMS)**

### **2.1 What is an Object-Oriented Database?**

An Object-Oriented Database Management System (OODBMS) integrates object-oriented programming principles with database technologies. Unlike relational systems that store data in rows and tables, OODBMS stores data as objects—similar to how programming languages like Java or Python represent them.

It extends the capabilities of databases by incorporating concepts such as **encapsulation**, **inheritance**, and **polymorphism** directly into the data model.

---

### **2.2 Core Features of OODBMS**

#### **Encapsulation**

Encapsulation is the bundling of data (attributes) and methods (operations) into a single unit called an object. In OODBMS:

- Data and associated behavior (methods) are stored together.  
- This leads to better abstraction and modularity.  
- Encapsulation hides implementation details from users or other systems, exposing only relevant interfaces.

For example, a Student object might include the name, ID, and a method `calculateGPA()` that operates on the data. This combination reflects real-world entities more naturally than relational schemas.

#### **Inheritance**

Inheritance allows one class (subclass) to inherit attributes and methods from another (superclass). In an OODBMS:

- Classes can be organized in hierarchies.  
- Subclasses extend or override methods/attributes of their parents.  
- This promotes reusability and reduces redundancy.

For example, a `GraduateStudent` class might inherit from a general `Student` class, while adding specific attributes such as `thesisTopic`.

---

### **2.3 Advantages of OODBMS**

- Seamless Integration with OOP Languages: OODBMS allows persistence of objects created in object-oriented programs, eliminating the "impedance mismatch" problem.  
- Natural Data Modeling: Ideal for applications that manage complex data, such as CAD/CAM, multimedia, simulation, and engineering databases.  
- Support for Complex Objects: Nested objects, recursive structures, and many-to-many relationships are better handled.

---

### **2.4 Disadvantages of OODBMS**

- Limited Standardization: Unlike SQL in RDBMS, there's less standardization in OODBMS query languages and interfaces.  
- Smaller Ecosystem: Fewer tools and community resources compared to relational systems.  
- Complexity: Inheritance and encapsulation introduce additional layers of abstraction that can be difficult for new users.

---

## **3. Deductive Databases**

### **3.1 Introduction**

A deductive database is a logic-based database system that extends the relational model by enabling **inference**—the ability to derive new facts from existing data using rules. It is particularly useful in knowledge-based systems and artificial intelligence applications.

---

### **3.2 Datalog: The Query Language of Deductive Databases**

**Datalog** is a declarative logic programming language, syntactically similar to Prolog but tailored for database queries.

Key elements:

- **Facts**: Basic assertions about data  
- **Rules**: Logical inferences based on other facts  
- **Queries**: Questions to ask the system based on the rules and facts

---

### **3.3 Features of Deductive Databases**

- Rule-based Querying: Users define rules to express complex relationships and dependencies.  
- Recursive Queries: Datalog naturally supports recursion, unlike standard SQL.  
- Non-Procedural Semantics: The logic expresses what to retrieve, not how to retrieve it.

---

### **3.4 Advantages**

- Powerful Reasoning: Can derive implicit facts from explicit data.  
- Good for Hierarchical Data: Recursive queries are efficient.  
- Formal Semantics: Provides a solid mathematical foundation for query optimization.

---

### **3.5 Limitations**

- Performance Issues: Recursive queries can be computationally intensive.  
- Limited Commercial Use: Not widely adopted in mainstream enterprise environments.

---

## **4. XML Databases**

### **4.1 What is XML?**

XML (eXtensible Markup Language) is a standard format for representing structured and semi-structured data in a hierarchical manner. XML is widely used in web services, configuration files, data interchange, and document storage.

An XML database is a data persistence system optimized for storing, querying, and managing XML data.

---

### **4.2 Structure of XML**

XML documents consist of elements nested within each other, with tags defining their structure.

This naturally models hierarchical data better than flat tables in relational systems.

---

### **4.3 XML Schema**

An XML Schema defines the structure and constraints of an XML document:

- Declares allowable elements and attributes  
- Specifies data types (string, integer, date, etc.)  
- Defines element ordering and hierarchy

Schemas ensure XML data adheres to a specific format, improving data validation and interoperability.

---

### **4.4 Querying XML Databases**

To retrieve or manipulate data in XML databases, two key technologies are used:

#### **XPath (XML Path Language)**

- A language for navigating XML documents using path expressions  
- Selects nodes or node-sets based on attributes, tags, or hierarchy  
- Example: `/university/student/name` selects all `name` elements under `student` elements inside `university`

#### **XQuery**

- A powerful query language for XML, combining XPath and FLWOR expressions (For-Let-Where-Order-Return)  
- Enables filtering, sorting, and transformation of XML data  
- Example: `for $x in /students/student where $x/age > 21 return $x/name` returns names of students older than 21

---

### **4.5 Types of XML Databases**

- **Native XML Databases**: Store XML documents directly (e.g., BaseX, eXist-db)  
- **XML-enabled Relational Databases**: Store XML as BLOBs or shredded into relational schemas (e.g., Oracle XML DB, SQL Server XML support)

---

### **4.6 Advantages of XML Databases**

- Flexible Schema: Accommodates evolving data models  
- Hierarchical Representation: Ideal for representing nested structures  
- Platform Independent: Widely supported across languages and systems

---

### **4.7 Limitations**

- Slower than RDBMS: Especially for flat, highly structured data  
- Complex Queries: XPath/XQuery can be difficult for new users  
- Redundant Data: Verbose markup increases storage size

### **Key Comparison Points**

| Feature           | Relational DBMS                  | Object-Oriented DBMS                        |

| Query Language    | SQL                              | Object Query Language (OQL)                 |
