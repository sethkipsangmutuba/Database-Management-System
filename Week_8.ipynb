# Week 8: Data Warehousing, OLAP & Analytics

### 1. Introduction
Modern organizations generate vast volumes of data through various operational systems such as sales, inventory, and customer relationship management. However, this transactional data is not well-suited for strategic analysis or decision-making. To bridge this gap, data warehousing and OLAP (Online Analytical Processing) systems are used to provide historical insights, trend analysis, and high-level summaries from large datasets. This lecture explores the architecture of data warehouses, OLAP modeling, schema design, and analytical operations.

### 2. What is a Data Warehouse?
A Data Warehouse (DW) is a centralized, integrated repository designed specifically for query and analysis. It consolidates data from different sources, transforms it into a consistent format, and stores it in a way optimized for analytical querying, not transaction processing.

According to Bill Inmon, a leading authority in the field, a data warehouse is:

> "Subject-oriented, integrated, time-variant, and non-volatile collection of data in support of management’s decision-making process."

#### 2.1 Key Characteristics
- **Subject-Oriented**: Organized around key business domains (e.g., customers, products, sales).
- **Integrated**: Data is cleaned and standardized across sources.
- **Time-Variant**: Supports historical data storage, often in snapshots.
- **Non-Volatile**: Once loaded, data is stable and only updated through batch refreshes.

### 3. Data Warehouse Architecture
A typical data warehouse system is built around a layered architecture:

#### 3.1 Source Systems
These include various OLTP (Online Transaction Processing) systems such as POS terminals, HRM systems, ERPs, and CRMs that generate operational data.

#### 3.2 ETL Layer (Extract, Transform, Load)
This component extracts data from source systems, transforms it into a common schema, and loads it into the warehouse. The process often includes data cleaning, deduplication, type conversion, aggregation, and enrichment.

#### 3.3 Data Storage Layer
This is the actual database system where the cleaned and integrated data is stored. It is typically organized into fact and dimension tables for analytical querying.

#### 3.4 Data Marts
These are subsets of the data warehouse tailored to specific departments or business units (e.g., sales, marketing).

#### 3.5 Presentation/Access Layer
Provides interfaces for end-users, such as dashboards, reports, ad-hoc query tools, and OLAP systems.

### 4. Schema Design in Data Warehousing
Designing a data warehouse schema is fundamental for supporting analytical queries efficiently. The most common schemas are **Star Schema** and **Snowflake Schema**.

#### 4.1 Star Schema
In a star schema, a central fact table is surrounded by dimension tables. The fact table contains numeric measures (e.g., sales, profit), and dimension tables contain attributes for filtering and grouping (e.g., customer name, product category, date).

**Advantages:**
- Simplified queries
- Faster performance due to denormalized structure

#### 4.2 Snowflake Schema
A snowflake schema normalizes the dimension tables into multiple related sub-tables. It is more normalized than the star schema.

**Advantages:**
- Reduces data redundancy
- More space-efficient

**Disadvantages:**
- Increases query complexity due to additional joins
- Potentially slower performance compared to star schema

### 5. Online Analytical Processing (OLAP)
OLAP systems allow users to analyze data across multiple dimensions interactively and efficiently. OLAP transforms raw data into useful insights through multi-dimensional structures known as OLAP cubes.

#### 5.1 OLAP Cubes
An OLAP cube organizes data into dimensions and measures:
- **Dimensions**: Categories for analysis (e.g., time, product, location)
- **Measures**: Quantitative data (e.g., sales, profits)

#### 5.2 OLAP Operations
OLAP provides powerful analytical operations that facilitate complex queries:

- **Roll-Up**: Aggregates data by climbing up a dimension hierarchy (e.g., from days to months).
- **Drill-Down**: Moves from summary data to more detailed levels (e.g., from years to quarters to months).
- **Slice**: Filters the cube by selecting a single dimension value (e.g., sales in 2022).
- **Dice**: Filters the cube by selecting specific values from multiple dimensions (e.g., sales of electronics in Nairobi during Q1).
- **Pivot (Rotate)**: Reorients the dimensions to view data from different perspectives.

These operations enable dynamic, exploratory analysis for end-users.

### 6. ETL Pipeline Design
An efficient ETL pipeline is the backbone of data warehousing. It consists of three phases:

- **Extraction**: Retrieves raw data from various operational systems and external sources.
- **Transformation**: Converts extracted data into a unified format, applying business rules, resolving conflicts, and performing data cleansing.
- **Loading**: Inserts transformed data into the data warehouse or data mart, often in batch mode.

ETL tools must handle data integrity, error logging, schema changes, and large volumes of data efficiently. Scheduling and automation of ETL tasks ensure regular updates to the warehouse.

### 7. Materialized Views

#### 7.1 Concept
A materialized view is a physical copy of query results stored in the database. Unlike regular views, which are computed on-demand, materialized views are precomputed and stored to optimize performance.

#### 7.2 Importance in OLAP
Materialized views are used in OLAP systems to:
- Improve response time for complex aggregations
- Avoid recalculating commonly requested summaries
- Support faster roll-up and drill-down operations

These views are especially helpful in systems where response time is critical, such as executive dashboards and interactive business intelligence applications.

### 8. Data Analytics and Business Intelligence
Data warehousing is closely linked to analytics and business intelligence (BI). With historical data stored in the warehouse, organizations can conduct a variety of analytical operations to support strategic decision-making.

#### 8.1 Types of Data Analytics
- **Descriptive Analytics**: Explains what has happened (e.g., monthly sales reports).
- **Diagnostic Analytics**: Explains why something happened (e.g., identifying reasons for sales drop).
- **Predictive Analytics**: Forecasts future trends using historical data (e.g., sales forecasting).
- **Prescriptive Analytics**: Recommends actions (e.g., inventory optimization).

#### 8.2 Business Intelligence Integration
BI tools sit on top of the data warehouse and allow users to:
- Create dashboards
- Generate real-time reports
- Monitor KPIs (Key Performance Indicators)
- Visualize trends and patterns

These tools provide a user-friendly interface for decision-makers to derive actionable insights.

```python
# ===============================================
# Week 8: Data Warehousing, OLAP & Analytics
# ===============================================

# === Objectives ===
# - Understand data warehousing architecture
# - Design star/snowflake schemas
# - Explore OLAP cubes and analytical operations
# - Simulate ETL pipelines and materialized views
# - Perform business analytics on sample data

# === Topics Covered ===
# - ETL, OLAP, roll-up, drill-down, slice, dice, pivot
# - Star/Snowflake schema
# - Analytics (descriptive, predictive)
# - Materialized views

# === Activities ===
# - Lab: Build mini data mart, simulate OLAP queries
# - Assignment: Retail schema design + OLAP strategy

# ---------------------------------------------
# Section 1: Simulate ETL - Extract, Transform, Load
# ---------------------------------------------
print("=== Section 1: ETL Simulation ===")

import pandas as pd

# Extraction: Raw operational data
raw_sales = pd.DataFrame({
    'ProductID': [101, 101, 102, 103, 102],
    'CustomerID': [1, 2, 1, 3, 2],
    'Region': ['Nairobi', 'Mombasa', 'Nairobi', 'Kisumu', 'Mombasa'],
    'Date': ['2023-01-05', '2023-01-05', '2023-01-06', '2023-01-06', '2023-01-07'],
    'Quantity': [3, 1, 2, 4, 5],
    'Price': [1000, 1000, 1500, 2000, 1500]
})

# Transformation: Add Total column, convert types
raw_sales['Date'] = pd.to_datetime(raw_sales['Date'])
raw_sales['Total'] = raw_sales['Quantity'] * raw_sales['Price']

# Loading: Load into warehouse (fact table)
fact_sales = raw_sales.copy()

print("Fact Table - Sales:")
print(fact_sales)

# ---------------------------------------------
# Section 2: Define Dimension Tables (Star Schema)
# ---------------------------------------------
print("\n=== Section 2: Dimension Tables ===")

dim_product = pd.DataFrame({
    'ProductID': [101, 102, 103],
    'ProductName': ['Laptop', 'Smartphone', 'Tablet'],
    'Category': ['Electronics', 'Electronics', 'Electronics']
})

dim_customer = pd.DataFrame({
    'CustomerID': [1, 2, 3],
    'CustomerName': ['Alice', 'Bob', 'Carol'],
    'Segment': ['Retail', 'Corporate', 'Retail']
})

dim_time = fact_sales[['Date']].drop_duplicates()
dim_time['Year'] = dim_time['Date'].dt.year
dim_time['Month'] = dim_time['Date'].dt.month
dim_time['Day'] = dim_time['Date'].dt.day

print("Dimension - Product:")
print(dim_product)

print("\nDimension - Customer:")
print(dim_customer)

print("\nDimension - Time:")
print(dim_time)

# ---------------------------------------------
# Section 3: OLAP Cube Simulation
# ---------------------------------------------
print("\n=== Section 3: OLAP Cube Simulation ===")

# Merge for cube
cube = fact_sales.merge(dim_product, on='ProductID')\
                 .merge(dim_customer, on='CustomerID')\
                 .merge(dim_time, on='Date')

print("OLAP Cube Sample:")
print(cube[['Date', 'ProductName', 'CustomerName', 'Region', 'Quantity', 'Total']])

# ---------------------------------------------
# Section 4: OLAP Operations
# ---------------------------------------------
print("\n=== Section 4: OLAP Operations ===")

# Roll-Up: Total sales by month
rollup = cube.groupby(['Year', 'Month'])['Total'].sum().reset_index()
print("Roll-Up (Total Sales by Month):")
print(rollup)

# Drill-Down: From month to day level
drilldown = cube.groupby(['Year', 'Month', 'Day'])['Total'].sum().reset_index()
print("\nDrill-Down (Sales by Day):")
print(drilldown)

# Slice: Filter by Region = Mombasa
slice_mombasa = cube[cube['Region'] == 'Mombasa']
print("\nSlice (Sales in Mombasa):")
print(slice_mombasa[['Date', 'ProductName', 'CustomerName', 'Total']])

# Dice: Filter sales for Smartphone in Nairobi
dice = cube[(cube['ProductName'] == 'Smartphone') & (cube['Region'] == 'Nairobi')]
print("\nDice (Smartphone sales in Nairobi):")
print(dice[['Date', 'CustomerName', 'Total']])

# Pivot: Sales by Product and Region
pivot = cube.pivot_table(index='ProductName', columns='Region', values='Total', aggfunc='sum')
print("\nPivot (Product vs Region - Total Sales):")
print(pivot.fillna(0))

# ---------------------------------------------
# Section 5: Materialized View Simulation
# ---------------------------------------------
print("\n=== Section 5: Materialized View (Precomputed Aggregates) ===")

# Simulate by storing precomputed summaries
materialized_sales_summary = cube.groupby(['ProductName', 'Region'])['Total'].sum().reset_index()
print("Materialized View: Product x Region Total Sales")
print(materialized_sales_summary)

# ---------------------------------------------
# Section 6: Business Analytics Summary
# ---------------------------------------------
print("\n=== Section 6: Business Analytics ===")

# Descriptive: Total revenue
print("Total Revenue:", cube['Total'].sum())

# Diagnostic: Which region spent most?
top_region = cube.groupby('Region')['Total'].sum().idxmax()
print("Top Spending Region:", top_region)

# Predictive (Simple Forecast): Linear trend for monthly sales (not actual model)
monthly = cube.groupby(['Year', 'Month'])['Total'].sum().reset_index()
monthly['NextMonthForecast'] = monthly['Total'].shift(-1).fillna(method='ffill')
print("\nSales Forecast (Dummy):")
print(monthly)

# ---------------------------------------------
# Section 7: Summary & Takeaways
# ---------------------------------------------
print("\n=== Section 7: Summary ===")
print("""
✔ Data Warehouse modeled using fact and dimension tables (star schema).
✔ Simulated ETL pipeline processed and loaded operational data.
✔ OLAP cube supported roll-up, drill-down, slice, dice, and pivot queries.
✔ Materialized views enabled faster access to frequent summaries.
✔ Simple analytics demonstrated descriptive and predictive insights.
""")

print("\n=== Suggested Tools ===")
print("""
- PostgreSQL with OLAP extensions
- Apache Hive for big data warehousing
- Power BI / Tableau for business dashboards
- Apache Airflow for ETL automation
""")
```
