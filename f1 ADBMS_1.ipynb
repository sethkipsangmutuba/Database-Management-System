##a) Introduction to Databases — Summary

Databases are central to everyday life  used in banking, shopping, reservations, education, and more.

- Multimedia
- Maps
- Analytics
- Real-time systems

- University systems  
- E-commerce platforms (e.g., Amazon)  
- Banking and finance  
- Healthcare systems

- The **data** itself  
- The **DBMS software**  
- The **users and applications** that interact with the system

They vary in size:
- Small personal databases (e.g., a contacts list)
- Massive enterprise systems (e.g., billions of records)

Includes:

##b)  Key Characteristics of the Database Approach (vs. File Processing)

The database approach addresses the limitations of traditional file-based systems by offering centralized, consistent, and shared data management. Here's how it stands out:

---

###  1. Self-Describing System

- A DBMS stores both the **data** and its **structure** (metadata) in a catalog.
- Metadata includes:  
  - Table names  
  - Data types  
  - Constraints  
  - Relationships
- Unlike traditional systems where data definitions are hardcoded, DBMSs access this catalog dynamically, allowing **flexibility and reuse**.

---

###  2. Program-Data Independence

- Changes to data structure (e.g., adding a new field) **do not require changes** to application programs.
- Data definitions are stored **separately from code**, enabling smoother updates and **minimal disruption**.

---

###  3. Data Abstraction

- Users interact with **logical representations** of data (e.g., `STUDENT(Name, ID, Major)`), not with physical storage.
- This hides complexity like file structure and memory layout.
- Higher-level models (e.g., object-relational) may include functions such as `CALCULATE_GPA()` as part of the abstraction.

---

###  4. Multiple Views

- A DBMS supports **customized views** for different users or applications.
- Example:  
  - One user sees only student transcripts  
  - Another user checks course prerequisites
- Views can:
  - Hide irrelevant data  
  - Display derived (virtual) data  
  - Improve **usability and security**

---

###  5. Data Sharing & Multiuser Access

- A DBMS allows **multiple users** to access data **concurrently** while maintaining consistency.
- **Concurrency control** prevents conflicts during simultaneous operations (e.g., booking the same seat).
- Ensures **ACID properties** for transactions:
  - **Atomicity**: All-or-nothing execution
  - **Isolation**: Each transaction executes as if it's the only one  
  - *(Other ACID properties covered in detail later)*

##c)  Advantages of the DBMS Approach

The database approach offers numerous benefits over traditional file-based systems. Below is a summary of its core advantages:

---

###  Reduced Data Redundancy

- **Centralized databases** eliminate unnecessary duplication.
- Saves storage and avoids inconsistencies across departments.
- **Controlled redundancy** (for performance) is managed through integrity rules within the DBMS.

---

###  Access Control & Security

- Ensures that **only authorized users** can access or modify specific data.
- Access is controlled using:
  - User accounts
  - Passwords
  - Privilege settings

---

###  Persistent Storage of Program Objects

- DBMSs can **store complex data objects** from programming languages like C++ or Java.
- Resolves the **impedance mismatch** between application code and data storage.

---

###  Efficient Query Processing

- Uses advanced techniques such as:
  - Indexing
  - Caching
  - Query optimization
- Minimizes disk access and enhances performance.

---

###  Backup and Recovery

- Provides **automated tools** for:
  - Data backup
  - Recovery after crashes or system failures

---

###  Multiple User Interfaces

- Supports a wide range of access methods:
  - SQL command-line for analysts
  - GUIs for non-technical users
  - APIs for developers
  - Web-based interfaces for remote access

---

###  Support for Complex Data Relationships

- Handles various relationships such as:
  - One-to-many
  - Many-to-many
- Enables relational integrity across tables and entities.

---

###  Integrity Constraints Enforcement

- Ensures data accuracy using rules like:
  - Data types
  - Uniqueness
  - Referential integrity
- Constraints are defined during the design phase.

---

###  Rule-Based Automation (Triggers & Procedures)

- Enables automatic actions through:
  - **Triggers**: Auto-execute when conditions are met
  - **Stored Procedures**: Encapsulated business logic
- Supports decision-making and workflow automation.

---

### Additional Benefits

- **Standardization**: Consistent data definitions and formats across the organization.
- **Faster Development**: Reuse of existing components and DBMS tools accelerates development.
- **Flexibility**: Schema changes can be made without disrupting applications.
- **Real-Time Updates**: Changes are instantly visible to all users.
- **Cost Efficiency**: Reduces duplication and lowers infrastructure/maintenance costs.


## d) A Brief History of Database Applications

---

###  Early Systems: Hierarchical & Network Models

- Used in large organizations like **banks, hospitals, and universities**.
- Data was tightly coupled with **physical storage layout**, reducing flexibility.
- Required **low-level programming** for queries and lacked abstraction.
  
---

###  Relational Databases: A Paradigm Shift

- Introduced in the **1970s** to **separate logical data representation** from physical storage.
- Enabled **high-level querying using SQL**.
- Improved **flexibility**, usability, and **data independence**.
- Became **dominant in the 1980s** due to:
  - Indexing
  - Query optimization
  - Better performance

---

###  Object-Oriented Databases (OODBs)

- Emerged in the **1980s** to support **complex and structured data**.
- Incorporated object-oriented features:
  - **Encapsulation**
  - **Inheritance**
- Adoption was limited due to:
  - Complexity
  - Lack of standards
- Led to the creation of **Object-Relational DBMSs (ORDBMSs)** — hybrids between relational and object-oriented systems.

---

###  Web & E-commerce Integration

- Rise of the **web** enabled **dynamic, database-driven websites**.
- **XML** became the standard for **web-based data interchange**.
- Enabled **seamless integration** and data sharing across platforms using web technologies.

---

###  Advanced Applications & Extended DBMS Features

Modern demands led to new uses and features in DBMSs, including:

#### Applications:
- **Scientific research** (e.g., genomics, physics experiments)
- **Multimedia** (images, video, audio)
- **Data mining** (pattern discovery)
- **Geospatial systems** (maps, GPS)
- **Time series analysis** (finance, economics)

#### Enhancements:
- Support for **new data types** and operations
- Specialized indexing for non-traditional data
- Optional DBMS **modules** (e.g., spatial, temporal data)
- Enterprise systems (e.g., **ERP**, **CRM**) built on DBMS back-ends and accessed via **web interfaces**

---

###  Databases vs. Information Retrieval (IR)

| Aspect                  | Databases                                | Information Retrieval (IR)              |
|------------------------|-------------------------------------------|----------------------------------------|
| **Focus**              | Structured data (tables, fields)          | Unstructured data (documents, text)    |
| **Query Method**       | SQL and precise queries                   | Keyword search and document ranking    |
| **Use Cases**          | Business apps, transactions               | Web search, digital libraries          |

- Modern systems often **combine both approaches**, especially on the web.

##e) When Not to Use a DBMS

While DBMSs offer many advantages, they are not always the best fit. In certain cases, their **overhead** may outweigh the benefits.

---

###  Key Drawbacks of Using a DBMS

- **High initial costs**:
  - Licensing for software
  - Investment in hardware
  - Staff training
- **Performance overhead** due to:
  - General-purpose architecture
  - Security, concurrency, and recovery processes
- **Additional processing** for:
  - Integrity enforcement
  - User management
  - Query optimization

---

###  Situations Where File-Based Systems May Be Preferable

- **Simple applications** with fixed, unchanging requirements
- **Real-time systems** where ultra-fast response is critical (DBMS latency is unacceptable)
- **Embedded systems** with limited storage and processing power
- **Single-user environments** where multi-user concurrency is unnecessary

---

###  Real-World Examples (Non-DBMS Use)

- **CAD tools**: Use custom file formats optimized for performance in 3D modeling
- **Telecom switches**: Require extremely fast hierarchical data access
- **GIS platforms**: May use spatial data structures not efficiently supported by general-purpose DBMSs

    - Performs core database functions.

## b) Data Models, Schemas, and Instances

### Data Abstraction

- Hides physical storage details.
- Shows users only relevant features of the data.

### What is a Data Model?

- Defines the **structure** of data: types, relationships, constraints.
- Defines **operations** on data: retrieval, update.
- May include **behavior** (operations tied to data objects), especially in object-oriented and object-relational models.

### Categories of Data Models

| Type                     | Description                                | Examples                         |
|--------------------------|--------------------------------------------|---------------------------------|
| **Conceptual (High-level)**    | Close to user’s view of data (entities, attributes, relationships) | Entity-Relationship (ER) model   |
| **Representational (Implementation)** | Closer to actual storage but user-friendly | Relational, Hierarchical, Network models |
| **Physical**              | Details of actual data storage (records, files, indexes) | File structures, access paths   |

### Schemas and Instances

- **Schema**: Database design or structure, rarely changes. (Metadata)
- **Instance**: Actual data stored at a point in time, frequently changes.
- DBMS stores the schema and enforces that data instances conform to it.
- **Schema evolution** occurs when database design changes to meet new requirements.

---

##  Three-Schema Architecture and Data Independence

### Three Levels of Schema

| Schema Level        | Description                                   |
|---------------------|-----------------------------------------------|
| **Internal Schema**  | Physical storage details (file organization, indexing) |
| **Conceptual Schema**| Logical structure of entire database (entities, relationships) |
| **External Schemas (Views)** | User-specific views that hide irrelevant data |

- This architecture **isolates applications** from physical data storage.
- Supports **multiple user views** customized for different needs.
- **Mappings** translate queries and data between levels (adds flexibility, some overhead).

### Data Independence

- **Logical Data Independence**: Ability to change the conceptual schema (e.g., add/remove fields) without affecting external schemas or application programs.
  - Difficult to achieve but critical for adaptability.
- **Physical Data Independence**: Ability to change internal schema (e.g., reorganize files, add indexes) without affecting conceptual schema or applications.
  - Easier and more common.

The **three-schema architecture** facilitates both by isolating changes to one layer and adjusting mappings accordingly.

#c) Database Languages and Interfaces

Databases serve many kinds of users, each requiring different ways to interact with the DBMS. To accommodate this, DBMSs provide specialized languages and interfaces tailored to user needs.

---

## DBMS Languages

- After designing a database and choosing a DBMS, the first task is to **define database schemas** (conceptual and internal) and their mappings.

- **Data Definition Language (DDL):**  
  Used to define schemas and store descriptions in the system catalog.

- In strict-level systems:  
  - DDL defines the **conceptual schema** (logical structure).  
  - **Storage Definition Language (SDL)** defines the **internal schema** (physical storage).  

- Most modern relational DBMSs **do not use separate SDL**; storage details are handled via functions or DBA-managed parameters.

- **View Definition Language (VDL):**  
  Ideally defines user views (external schemas) and their mappings, but often DDL doubles as VDL.

- **SQL** acts as a combined DDL, VDL, and DML in relational DBMSs.

---

### Data Manipulation Language (DML)

- Used by users to **manipulate data** (retrieve, insert, delete, update).

- Types of DML:
  - **High-level (nonprocedural) DML:**  
    - Specifies *what* data to retrieve, not *how* to retrieve it.  
    - Concise, usable interactively or embedded in programming languages.  
    - Example: SQL.
  - **Low-level (procedural) DML:**  
    - Requires detailed steps to access data (record-by-record).  
    - Embedded in programming languages for fine-grained control.

- When embedded, the **host programming language** manages logic; DML acts as a data sublanguage.

- **Query languages:** Standalone, high-level DML used interactively.

- User types & typical interaction methods:
  - Casual users: Use high-level query languages.
  - Programmers: Use embedded DML.
  - Naive and parametric users: Use user-friendly interfaces rather than direct query languages.

---

## DBMS Interfaces

DBMSs provide a variety of user-friendly interfaces to accommodate different user skill levels:

- **Menu-Based Interfaces:**  
  Users select options from menus to build queries step-by-step without knowing command syntax. Common on web platforms.

- **Forms-Based Interfaces:**  
  Present forms to fill in fields for data insertion or searching. Designed for naive users performing routine tasks (e.g., Oracle Forms, SQL*Forms).

- **Graphical User Interfaces (GUIs):**  
  Display schemas graphically, allowing users to point and click to construct queries. Often combine menus and forms for ease of use.

- **Natural Language Interfaces:**  
  Allow queries in English or other natural languages, which the system attempts to interpret and convert into formal queries. Common in web search engines but still limited.

- **Speech Input/Output Interfaces:**  
  Use speech recognition for querying in limited domains (e.g., flight info, phone directories). Speech can also be used for output.

- **Interfaces for Parametric Users:**  
  Designed for users (e.g., bank tellers) who perform a small set of predefined tasks repeatedly. Provide quick access via function keys or shortcuts.

- **Interfaces for Database Administrators (DBAs):**  
  Include privileged commands for administrative functions such as creating users, setting permissions, changing schemas, and reorganizing storage. Restricted to DBAs.

#d) The Database System Environment

A **DBMS** (Database Management System) is a complex software system consisting of multiple components that manage data storage, transaction processing, and user interaction.

---

##  DBMS Component Modules

> **Figure 2.3** (not shown here) illustrates two parts of a typical DBMS architecture:
- **Top:** Users and their interfaces
- **Bottom:** Internal DBMS components for data storage and transaction processing

###  Storage
- The **database** and **DBMS catalog** are stored on disk.
- Disk access is controlled either by:
  - The **Operating System (OS)**, or
  - The **DBMS’s buffer management module** to optimize performance.

---

###  Top Part: User Interfaces and Access

#### Users:

- **DBA Staff**: Define schemas, tune the system via DDL and privileged commands.
- **Casual Users**: Use interactive query interfaces to formulate queries.
- **Application Programmers**: Write host language programs (e.g., Java, C, C++) embedding DML commands.
- **Parametric Users**: Use canned transactions by supplying parameters (e.g., bank withdrawal).

#### Components:

- **DDL Compiler**:
  - Processes schema definitions.
  - Stores metadata in the system catalog (file info, data types, storage details, constraints, mappings).

- **Interactive Query Interface**:
  - Parses and validates queries.
  - Compiles them into internal formats.

- **Query Optimizer**:
  - Rearranges operations, removes redundancies.
  - Selects optimal algorithms and indexes using catalog statistics.

- **Precompiler**:
  - Extracts embedded DML commands from host programs.
  - Sends them to the DML compiler.

- **DML Compiler**:
  - Compiles DML commands into object code.
  - Links them with host programs to create executable canned transactions.

---

###  Lower Part: Runtime Execution

- **Runtime Database Processor**: Executes:
  - Privileged commands
  - Query plans
  - Canned transactions with parameters

#### Works with:

- **System Catalog**: May update statistics.
- **Stored Data Manager**: Handles data access via OS I/O.
- **Buffer Management**: Controlled by DBMS or OS.
- **Concurrency Control & Backup/Recovery Modules**: Manage transactions.

---

###  Client/Server Model

- **Client Programs**: Run on user machines.
- **Database Server**: Stores data, processes transactions.
- **Application Server** (optional): Acts as an intermediary.

> The DBMS interfaces with:
- OS for disk access
- Programming language compilers
- Network communication in distributed environments

---

##  Database System Utilities

Utilities help DBAs manage the database:

- **Loading Utility**: Loads external data files, auto-converts formats.
- **Backup Utility**: Creates full/incremental backups for recovery.
- **Storage Reorganization Utility**: Reorganizes files, creates access paths to improve performance.
- **Performance Monitoring Utility**: Collects statistics to guide optimization decisions (e.g., indexing, file layout).
- **Other Utilities**: Sorting, compression, user monitoring, network interfacing.

---

##  Tools, Application Environments, and Communication Facilities

###  CASE Tools

- Help in database design and development (e.g., ER diagrams, schema generation).

###  Data Dictionary / Information Repository

- Extended catalogs that store:
  - Design decisions
  - Usage standards
  - Program metadata

Accessible to both users and DBAs.

###  Application Development Environments

- Tools like **PowerBuilder**, **JBuilder**:
  - Assist GUI design
  - Support application development
  - Enable database query/update

### Communication Software

- Allows remote access to databases via:
  - Internet
  - LANs
  - Phone lines
  - Satellites

> Fully integrated systems are known as **DB/DC systems**.

---

- Use LANs or other networks to **connect multiple machines**.

#e)  Centralized and Client/Server Architectures for DBMSs

Database systems can be organized based on how their components (like interfaces, applications, and data storage) are distributed across hardware. Two main architectural models dominate: **centralized** and **client/server**.

---

##  Centralized Architecture

- In early systems, **everything** — applications, user interfaces, and DBMS functions — ran on a **single mainframe computer**.
- Terminals connected to this machine were **"dumb" devices**, meaning:
  - They only **displayed output** and **sent input**.
- With the rise of **powerful and affordable PCs**, users started replacing terminals.
- Initially, the **DBMS still ran entirely on the central server**, even with PCs.
- Eventually, the **client-side computing power** was utilized, leading to **distributed and client/server architectures**.

---

##  Basic Client/Server Architecture

- This model **splits responsibilities** between clients and servers.

###  Clients:
- Typically **PCs or workstations**.
- Handle **user interfaces** and some **local processing**.

###  Servers:
- Offer **specialized services**, e.g.:
  - File storage
  - Printing
  - Database access

###  Workflow:
- Clients **send requests** (e.g., query, print).
- Servers **perform services** and **respond**.

> Multiple clients can share resources on a single server.

- This model gave rise to **database servers** (DBMS servers) that handle **client data requests**.

---



image.png

##  Two-Tier Client/Server Architecture for DBMSs

###  Structure:

- **Client Tier**:
  - Manages **user interface** and **application logic**.
- **Server Tier**:
  - Handles **data processing**, **transactions**, and **SQL execution**.

###  Communication:

- Clients use APIs like:
  - **ODBC** (Open Database Connectivity)
  - **JDBC** (Java Database Connectivity)

> These APIs allow SQL queries to be sent and results retrieved.

### Example:

- Most **Relational DBMSs (RDBMSs)** use this model.
- Server is often called a **query server** or **SQL server**.

### Object-Oriented DBMSs:

- Responsibilities may be shared more evenly:
  - Server: Manages **low-level storage** and **concurrency**.
  - Client: Handles **object structuring**.

---




##  Three-Tier (and n-Tier) Architectures

A third layer — the **application server** or **web server** — is introduced between the client and DBMS server.

###  Tier Breakdown:

- **Client Tier**:
  - Manages **user interface** (GUI, browsers, forms).
- **Application Tier**:
  - Processes **application logic** and **business rules** (e.g., validation, access control).
- **Database Tier**:
  - Handles **data storage** and **query execution**.

###  Advantages:

- **Enhanced Security**:
  - Application server can **validate requests** before they reach the database.
- **Greater Scalability**:
  - Layers can scale independently.
- **Modularity**:
  - Easier to manage and update specific layers.

> Widely used in **web applications** and **enterprise systems** (e.g., ERP, CRM).

###  n-Tier Architecture:

- In some systems, the architecture extends to **4 or 5 tiers**.
- Allows:
  - **Independent deployment and updates**
  - **Platform separation**

---

## Key Benefits of Modern Architectures

- **Improved Performance**:
  - Distributes processing across layers.
- **Increased Security**:
  - Layered access control and encryption.
- **Easier Maintenance & Scalability**:
  - Especially beneficial for **web-based and enterprise systems**.
- **Supports Multi-client Access**:
  - Efficient resource sharing across many users.

# Classification of Database Management Systems

DBMSs can be classified based on various criteria. This section outlines the major classifications, showing how systems differ in structure, design, and purpose.

---

## 1. By Data Model

The **data model** defines how data is structured and interacted with. This is the **primary classification** for DBMSs.

### a. Relational Model

- Most common and widely used in commercial DBMSs.
- Data is organized into **tables (relations)**.
- Uses **SQL** for data querying.

 **Examples**: MySQL, PostgreSQL, Oracle DB, MS SQL Server

---

### b. Object Model

- Structures data as **objects** with attributes and methods.
- Supports **inheritance** and **class hierarchies**.
- Less common in commercial use.
- Influenced later relational systems.

---

### c. Object-Relational Model

- Combines **relational tables** with **object-oriented features**.
- Supports **user-defined types**, inheritance, and **complex data structures**.

 **Example**: PostgreSQL (with object-relational support)

---

### d. Hierarchical Model

- Data organized in **tree-like parent-child structures**.
- Popular in early systems.

 **Example**: IBM IMS

- Still used in **government, healthcare, and banking** systems.

---

### e. Network Model

- Uses **graph structures** with **record types and set types**.
- Supports **many-to-many relationships**.

 **Examples**: IDMS, IMAGE, SUPRA  
 Based on the **CODASYL DBTG standard**

---

### f. XML Model

- Based on **hierarchical tree structures**.
- Popular for **web data exchange** and **semi-structured data**.
- **Native XML databases** exist.
- Many RDBMSs support **XML integration**.

---

## 2. By Number of Users

- **Single-user DBMSs**: Allow only one user at a time (often PC-based).
- **Multi-user DBMSs**: Support concurrent access by many users (most modern systems).

---

## 3. By Number of Sites (Centralized vs. Distributed)

### a. Centralized DBMS

- Entire DBMS and database reside on a **single computer**.
- Can still support **multiple users** via terminals or networks.

---

### b. Distributed DBMS (DDBMS)

- Database and DBMS software are distributed across **multiple networked locations**.

#### Types:

- **Homogeneous DDBMS**: All sites use the **same DBMS software**.
- **Heterogeneous DDBMS**: Sites use **different DBMSs**.

---

### c. Federated / Multidatabase Systems

- Use **middleware** to access **multiple autonomous and heterogeneous DBMSs**.
- Sites maintain **local autonomy**, but cooperate on **global queries**.

---

## 4. By Cost

### a. Open-Source / Free DBMSs

- Examples: **MySQL**, **PostgreSQL**
- Usually supported by **community contributors**.

### b. Commercial DBMSs

- Examples: **Oracle**, **MS SQL Server**, **IBM DB2**
- Offer **modular licensing** and **tiered features**.

 **Pricing models**:
- Site licenses (unlimited use per location)
- Per-user / per-seat licenses
- Cloud-based **subscription pricing**

---

## 5. By Access Path Support

Some DBMSs specialize in how data is stored and retrieved:

- **Inverted file systems**
- **Indexed access**
- **Hashed access paths**

> These enhance performance, especially in **search-intensive applications**.

---

## 6. By Purpose: General vs. Special-Purpose DBMSs

### a. General-Purpose

- Designed to support **a wide variety of applications**.

 **Examples**: PostgreSQL, Oracle

---

### b. Special-Purpose

Built for specific applications, such as:

- **Airline reservation systems**
- **Telecommunication systems**

These often fall under:

 **OLTP Systems** (Online Transaction Processing):  
Designed for **high concurrency** and **low latency**.

---

## Legacy Systems Overview

- **Network and Hierarchical DBMSs** are still in use in **critical infrastructures**.
- Tools like:
  - **DL/1** (Hierarchical)
  - **DBTG Network DML**  
  were influential in early DBMS development.

> These systems use **navigational, low-level programming**  
> rather than **set-based SQL queries**.

# Summary – Database System Concepts & Architecture

---

## 1. Data Models

Database models define **how data is structured and interpreted**:

- **Conceptual (High-level)**:  
  - Based on **entities and relationships**  
  -  Example: **ER Model**

- **Physical (Low-level)**:  
  - Describes **how data is stored on disk**

- **Representational**:  
  - Practical models used in real DBMSs  
  -  Examples: **Relational**, **Object-Oriented**

---

## 2. Schemas and Instances

- **Schema**:
  - The database’s structure — its **blueprint**
  - Changes **infrequently**

- **Instance**:
  - The **actual data** stored at a specific time
  - Changes **frequently**

---

## 3. Three-Schema Architecture

This architecture enables **abstraction** and **data independence**:

- **Internal Schema**:
  - Describes **physical storage details**

- **Conceptual Schema**:
  - Describes the **logical view** of the entire database

- **External Schemas**:
  - Define **customized views** for different users

 **Mappings** connect these levels  
 Supports **Logical** and **Physical Data Independence**

---

## 4. DBMS Languages

- **DDL (Data Definition Language)**:
  - Defines schemas, views, and sometimes storage

- **SDL (Storage Definition Language)**:
  - Often merged with DDL in modern systems

- **DML (Data Manipulation Language)**:
  - Used to **query** and **modify** data

   **High-level**: Declarative (e.g., SQL)  
   **Low-level**: Procedural (record-at-a-time)

---

## 5. Interfaces & Users

Various user types interact with the DBMS via different interfaces:

- **Types of Interfaces**:
  - Menu-based
  - Form-based
  - Graphical User Interfaces (GUI)
  - Application Programming Interfaces (APIs)
  - Natural language interfaces

- **User Roles**:
  - Casual users
  - Database Administrators (DBAs)
  - Application developers

---

## 6. System Architecture

- **Two-Tier**:
  - Client directly communicates with DB server

- **Three-Tier**:
  - Adds **application layer** between client and database

- **n-Tier**:
  - Common in **modern web apps**
  - Enhances **scalability** and **modularity**

---

DBMSs are classified based on various criteria:

###  Data Model:

- Hierarchical
- Network
- XML

###  Users:
- Single-user
- Multi-user

###  Sites:
- Centralized

###  Access Paths:
- Indexed
- Hashed
- Inverted file systems

###  Cost:

 **Licensing Models**:
