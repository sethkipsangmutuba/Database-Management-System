# Section 8: The Enhanced Entity-Relationship (EER) Model

---

## Overview

The **EER model** is an extension of the traditional **ER model**.

It was developed to model more **complex, real-world domains** such as:

- Engineering design (CAD/CAM)
- Telecommunications
- GIS
- Software systems

These domains require greater **semantic expressiveness** than traditional ER modeling offers.

---

## Why EER?

While the traditional ER model is good for **business-style databases**, modern applications have:

- More **complex relationships**
- **Hierarchies** and **inheritance**
- **Shared subtypes** and **categories**
- **Polymorphism-like** behavior

EER integrates concepts from:

- **Semantic modeling**
- **Object-oriented modeling**
- **AI knowledge representation**

---

## Key Enhancements in EER

The EER model adds the following advanced modeling constructs to the traditional ER model:

---

### 1. **Class/Subclass and Type Inheritance** *(Section 8.1)*

- Supports **entity hierarchies**
- Subclasses **inherit attributes and relationships** from superclasses

---

### 2. **Specialization and Generalization** *(Section 8.2)*

- Mechanisms for modeling **subset/superset** relationships
- Supports both:
  - **Top-down modeling** (specialization)
  - **Bottom-up modeling** (generalization)

---

### 3. **Constraints on Specialization/Generalization** *(Section 8.3)*

- **Disjoint** vs. **Overlapping** subclasses
- **Total** vs. **Partial** participation

---

### 4. **Union and Categories** *(Section 8.4)*

- **Category (union type)** allows modeling of a **subclass from multiple superclasses**
- Useful for modeling **shared entities** across hierarchies

---

### 5. **EER Example: UNIVERSITY Schema** *(Section 8.5)*

- A comprehensive **case study** showing full EER features in action
- Introduces **formal EER definitions**

---

### 6. **UML Class Diagrams** *(Section 8.6)*

- Discusses how **UML notation** represents **EER concepts**
- Continuation from **ER-to-UML mapping** in Chapter 7

---

### 7. **Semantic Modeling Foundations** *(Section 8.7)*

- Discusses abstractions such as:
  - **Classification**
  - **Aggregation**
  - **Generalization**
- These are **building blocks** for **semantic** and **object-oriented** models

# 8.1 Subclasses, Superclasses, and Inheritance — Lecture Notes

---

## What This Section Covers:

This section introduces the first major EER modeling feature:

- Subclass and superclass (also called subtype and supertype)
- Inheritance of attributes and relationships
- EER diagrammatic representation

---

## 1. Superclass/Subclass Relationship

In EER modeling:

- A **superclass** is a general entity type (e.g., `EMPLOYEE`)
- A **subclass** is a specialized version of the superclass (e.g., `SECRETARY`, `MANAGER`, `ENGINEER`)

Every **subclass entity** is also a member of the **superclass**.

**Example:**  
> “Joan Logano” is both an `EMPLOYEE` and a `SECRETARY`.

This is referred to as an **IS-A** relationship:

- `SECRETARY IS-A EMPLOYEE`
- `TECHNICIAN IS-A EMPLOYEE`

---

## 2.  Type Inheritance

**Inheritance** is the mechanism by which:

- A subclass **inherits** all attributes and relationships of its superclass.
- It can also define its **own (local)** attributes/relationships.

**Example:**

- `EMPLOYEE` has attributes: `Name`, `Ssn`, `Birth_date`, `Address`
- `ENGINEER` subclass adds: `Eng_type`, `Pay_scale`

🔸 An `ENGINEER` entity has both the attributes from `EMPLOYEE` and `ENGINEER`.

---

## 3.  Rules of Subclass Membership

- An entity **must be a member of the superclass** to be in a subclass.

- A superclass entity:
  - Can belong to **multiple subclasses** (e.g., `ENGINEER` and `SALARIED_EMPLOYEE`)
  - **Does not have to** belong to any subclass.

---

## 4.  EER Diagram Notation *(See Figure 8.1)*

Subclass relationships are diagrammed as follows:

- A **circle** labeled with the **specialization name** connects the superclass to its subclasses.
- A **line** connects the superclass to the circle, then branches out to all its subclasses.

🔸 This shows a **specialization** of the superclass (explained further in Section 8.2).

**Example (from Figure 8.1):**

- `EMPLOYEE` has 3 specializations:
  - `{SECRETARY, TECHNICIAN, ENGINEER}`
  - `{MANAGER}`
  - `{HOURLY_EMPLOYEE, SALARIED_EMPLOYEE}`

---

## 5.  Implications for Database Design

- Subclass entities can be modeled as **separate tables** linked by key to the superclass.
- When querying or managing entities, the **inheritance chain** should be respected.

---

##  Key Terminology

| **Term**        | **Meaning**                                            |
|-----------------|--------------------------------------------------------|
| Superclass      | General entity type (e.g., `EMPLOYEE`)                 |
| Subclass        | Specialized entity type (e.g., `TECHNICIAN`)           |
| IS-A            | Subclass-to-superclass relationship                    |
| Inheritance     | Subclass gains all properties (attributes + relationships) from superclass |

---


```python
# Base class: EMPLOYEE (superclass)
class Employee:
    def __init__(self, name, ssn, birth_date, address):
        self.name = name
        self.ssn = ssn
        self.birth_date = birth_date
        self.address = address

    def get_details(self):
        return f"Employee: {self.name}, SSN: {self.ssn}, Birth Date: {self.birth_date}, Address: {self.address}"


# Subclass: SECRETARY
class Secretary(Employee):
    def __init__(self, name, ssn, birth_date, address, typing_speed):
        super().__init__(name, ssn, birth_date, address)
        self.typing_speed = typing_speed  # words per minute

    def get_details(self):
        return f"{super().get_details()}, Typing Speed: {self.typing_speed} WPM"


# Subclass: ENGINEER
class Engineer(Employee):
    def __init__(self, name, ssn, birth_date, address, eng_type, pay_scale):
        super().__init__(name, ssn, birth_date, address)
        self.eng_type = eng_type
        self.pay_scale = pay_scale

    def get_details(self):
        return f"{super().get_details()}, Type: {self.eng_type}, Pay Scale: {self.pay_scale}"


# Subclass: MANAGER
class Manager(Employee):
    def __init__(self, name, ssn, birth_date, address, department):
        super().__init__(name, ssn, birth_date, address)
        self.department = department

    def get_details(self):
        return f"{super().get_details()}, Department: {self.department}"


# Subclass: HOURLY_EMPLOYEE
class HourlyEmployee(Employee):
    def __init__(self, name, ssn, birth_date, address, hourly_rate):
        super().__init__(name, ssn, birth_date, address)
        self.hourly_rate = hourly_rate

    def get_details(self):
        return f"{super().get_details()}, Hourly Rate: ${self.hourly_rate}/hr"


# Subclass: SALARIED_EMPLOYEE
class SalariedEmployee(Employee):
    def __init__(self, name, ssn, birth_date, address, annual_salary):
        super().__init__(name, ssn, birth_date, address)
        self.annual_salary = annual_salary

    def get_details(self):
        return f"{super().get_details()}, Annual Salary: ${self.annual_salary}"


# -----------------------------
# Example Usage
# -----------------------------
if __name__ == "__main__":
    # An Engineer (IS-A Employee)
    eng = Engineer("Alice Smith", "123-45-6789", "1985-06-12", "123 Main St", "Software", "Level 2")
    print(eng.get_details())

    # A Manager
    mgr = Manager("Bob Johnson", "111-22-3333", "1970-03-22", "456 Elm St", "Engineering")
    print(mgr.get_details())

    # A Secretary
    sec = Secretary("Carol White", "999-88-7777", "1990-10-01", "789 Oak Ave", 85)
    print(sec.get_details())

    # A Salaried Employee
    sal_emp = SalariedEmployee("Dana Grey", "444-55-6666", "1992-01-30", "321 Maple Rd", 90000)
    print(sal_emp.get_details())

    # A Hourly Employee
    hour_emp = HourlyEmployee("Eric Blue", "222-33-4444", "1995-08-15", "654 Pine Blvd", 25.5)
    print(hour_emp.get_details())
```

#### 8.2 Specialization and Generalization

---

## • a. Specialization

### 🔹 Definition:
**Specialization** is the process of defining **subclasses** from a **superclass** based on **distinctive characteristics**.

---

###  Example:

From `EMPLOYEE`, we can create subclasses:

- `{SECRETARY, ENGINEER, TECHNICIAN}` → based on **job type**
- `{SALARIED_EMPLOYEE, HOURLY_EMPLOYEE}` → based on **method of pay**

---

###  Diagram Representation (EER):

- Subclasses are **linked to a circle** representing specialization.
- This circle connects to the **superclass**.
- Subset symbols ($\subseteq$) indicate subclass direction.
- Subclass-specific attributes (like `TypingSpeed` for `SECRETARY`) are shown in the subclass rectangle.
- Subclasses may participate in relationships **not applicable** to the superclass.

**Example:**
- `HOURLY_EMPLOYEE` → `BELONGS_TO` → `TRADE_UNION`

---

###  Key Points:

- A real-world entity can appear in **both superclass and subclass**.
  - Ex: `e1` is both an `EMPLOYEE` and a `SECRETARY`.

- This resembles a **1:1 relationship** at the instance level, but the entity is the **same**, just playing a specialized role.

---

###  Reasons for Using Specialization:

1. Assign **attributes** to only some members of the superclass.
2. Define **relationships** involving only specific subsets of the superclass.

---

###  Summary – What Specialization Allows:

- Define a set of **subclasses** from a superclass.
- Add **specific (local) attributes** to subclasses.
- Define **relationship types** that apply only to specific subclasses.

---

## • b. Generalization

### 🔹 Definition:
**Generalization** is the **reverse** of specialization.  
It abstracts **common features** of multiple entity types into a **superclass**.

---

###  Example:

- `CAR` and `TRUCK` → generalized into `VEHICLE`

---

###  Diagram Representation:

- The **superclass (`VEHICLE`)** is connected via a **circle** to its subclasses (`CAR`, `TRUCK`).
- Some notations may use an **arrow pointing upward** to indicate **generalization**.

---

###  Notes:

- Generalization can be viewed as **specialization in reverse**.
- Diagrammatically, both **look the same** — the difference is in the **design intent**.

####  Directional Notation (in some methodologies):

- Arrow **upward** to superclass = **Generalization**
- Arrow **downward** to subclasses = **Specialization**

---

### Conclusion:

- Both **superclasses** and **subclasses** are types of entities.
- Both are represented as **rectangles** in EER diagrams.
- They help **model the real world more accurately** by organizing entities based on **shared** or **distinct** characteristics.

---


###Specialization of Employee

```python
# Superclass: Employee
class Employee:
    def __init__(self, emp_id, name):
        self.emp_id = emp_id
        self.name = name

    def describe(self):
        return f"Employee[ID: {self.emp_id}, Name: {self.name}]"


# Specialization: Secretary (with additional attribute)
class Secretary(Employee):
    def __init__(self, emp_id, name, typing_speed):
        super().__init__(emp_id, name)
        self.typing_speed = typing_speed

    def describe(self):
        return f"{super().describe()}, Typing Speed: {self.typing_speed} WPM"


# Specialization: Engineer (with additional attribute)
class Engineer(Employee):
    def __init__(self, emp_id, name, specialty):
        super().__init__(emp_id, name)
        self.specialty = specialty

    def describe(self):
        return f"{super().describe()}, Specialty: {self.specialty}"


# Specialization: Technician (no new attributes, just a type distinction)
class Technician(Employee):
    def describe(self):
        return f"{super().describe()}, Role: Technician"


# Specialization by Pay Method: Salaried vs Hourly
class SalariedEmployee(Employee):
    def __init__(self, emp_id, name, annual_salary):
        super().__init__(emp_id, name)
        self.annual_salary = annual_salary

    def describe(self):
        return f"{super().describe()}, Annual Salary: ${self.annual_salary}"


class HourlyEmployee(Employee):
    def __init__(self, emp_id, name, hourly_rate):
        super().__init__(emp_id, name)
        self.hourly_rate = hourly_rate

    def describe(self):
        return f"{super().describe()}, Hourly Rate: ${self.hourly_rate}/hr"
```

###Generalization Example: Vehicle ← Car, Truck

```python
# Subclasses: Car and Truck
class Car:
    def __init__(self, license_plate, model):
        self.license_plate = license_plate
        self.model = model

# Generalization: Vehicle
class Vehicle:
    def __init__(self, license_plate, vehicle_type):
        self.license_plate = license_plate
        self.vehicle_type = vehicle_type

    def describe(self):
        return f"Vehicle[Plate: {self.license_plate}, Type: {self.vehicle_type}]"
```

###Generalization Usage (Manual Aggregation)

```python
# Creating CAR and TRUCK entities
car = Vehicle("ABC123", "Car")
truck = Vehicle("XYZ789", "Truck")

# Generalized description
print(car.describe())
print(truck.describe())
```

###Example Execution

```python
if __name__ == "__main__":
    e1 = Secretary(101, "Alice", 85)
    e2 = Engineer(102, "Bob", "Mechanical")
    e3 = Technician(103, "Charlie")
    e4 = SalariedEmployee(104, "Dana", 95000)
    e5 = HourlyEmployee(105, "Eli", 30)

    # Print descriptions
    print(e1.describe())
    print(e2.describe())
    print(e3.describe())
    print(e4.describe())
    print(e5.describe())

    # Generalization demo
    car = Vehicle("CAR-123", "Car")
    truck = Vehicle("TRK-456", "Truck")
    print(car.describe())
    print(truck.describe())
```

# 8.3 Constraints and Characteristics of Specialization and Generalization Hierarchies

---

## • a. Constraints on Specialization and Generalization

These constraints apply to both specialization and generalization, though the examples below focus on specialization.

---

### i. Predicate-defined and Attribute-defined Subclasses

- **Predicate-defined subclass**: Defined by a condition on a superclass attribute.
  - Example:  
    `SECRETARY` subclass → condition: `Job_type = 'Secretary'`

- **Attribute-defined specialization**:  
  - All subclass definitions use the same **defining attribute**.  
  - Example:  
    Subclasses `{SECRETARY, TECHNICIAN, ENGINEER}` defined by `Job_type`.

- **User-defined subclass**:  
  - No formal condition.  
  - Users manually assign entities to subclasses.

---

### ii. Disjointness Constraint

- **Disjoint (symbol: d)**:
  - An entity can belong to **only one** subclass.
  - Enforced when using a **single-valued attribute**.

- **Overlapping (symbol: o)**:
  - An entity can belong to **multiple** subclasses.

---

### iii. Completeness Constraint

- **Total specialization** (double line):
  - Every entity **must** be in **at least one** subclass.

- **Partial specialization** (single line):
  - Some entities may **not** belong to **any** subclass.

---

### iv. Combined Constraints — Four Possibilities

1. **Disjoint, Total**  
2. **Disjoint, Partial**  
3. **Overlapping, Total**  
4. **Overlapping, Partial**

---

### v. Insertion and Deletion Rules

- **Delete from superclass** → Automatically deletes from all its subclasses.

- **Insert into superclass**:
  - If **predicate-defined** → entity is added to all matching subclasses.
  - If **total specialization** → must be added to **at least one** subclass.

---

## • b. Specialization and Generalization Hierarchies and Lattices

---

### i. Hierarchies vs. Lattices

- **Hierarchy**:
  - Each subclass has only **one** superclass.
  - Forms a **tree**.

- **Lattice**:
  - Subclass can have **multiple** superclasses (multiple inheritance).
  - Forms a **graph**.

- **Example**:  
  `ENGINEERING_MANAGER` is a subclass of both `ENGINEER` and `MANAGER` → **shared subclass** → lattice

---

### ii. Multiple Inheritance

- A shared subclass **inherits attributes from all** its superclasses.

- If an attribute is inherited through multiple paths, it **appears only once**.

- **Example**:  
  `STUDENT_ASSISTANT` inherits from both `STUDENT` and `EMPLOYEE`, which both inherit from `PERSON`.  
  → Attributes from `PERSON` appear only **once** in `STUDENT_ASSISTANT`.

---

### iii. Models Without Multiple Inheritance

- Some models **do not allow** multiple inheritance.

- Entities must belong to **only one leaf class**.

- All combinations of subclass memberships must be explicitly modeled.  
  Example for `PERSON` specialized into `{EMPLOYEE, ALUMNUS, STUDENT}`:

  - Must create 7 combinations:  
    `E`, `A`, `S`, `E_A`, `E_S`, `A_S`, `E_A_S`

---

## • c. Utilizing Specialization and Generalization in Schema Design

---

### i. Specialization: Top-Down Refinement

- Start with a general entity and create more specific subclasses.

- **UNIVERSITY Schema Example**:
  - Define `PERSON`
  - Specialize into `{EMPLOYEE, STUDENT, ALUMNUS}`
  - Further specialize into:
    - `FACULTY`, `STAFF`, `GRADUATE_STUDENT`, etc.
  - `STUDENT_ASSISTANT` is both a `STUDENT` and an `EMPLOYEE` → **lattice**

---

### ii. Generalization: Bottom-Up Synthesis

- Start with specific entity types and generalize shared features.

- **Example Path**:
  - Identify `RESEARCH_ASSISTANT`, `TEACHING_ASSISTANT`
  - Generalize → `STUDENT_ASSISTANT`
  - Generalize `{STAFF, FACULTY, STUDENT_ASSISTANT}` → `EMPLOYEE`
  - Generalize `{EMPLOYEE, STUDENT, ALUMNUS}` → `PERSON`

---

### iii. Practical Design Use

- In practice, both **specialization** and **generalization** are used together.

- Final structure (hierarchy/lattice) is the same; only the **design process** differs.

---


```python
# Base superclass
class Person:
    def __init__(self, pid, name, role=None):
        self.pid = pid
        self.name = name
        self.role = role  # Used for predicate-based specialization

    def describe(self):
        return f"Person[ID: {self.pid}, Name: {self.name}, Role: {self.role}]"

# Predicate-based specialization: subclasses inherit from Person

class Engineer(Person):
    def __init__(self, pid, name, eng_type):
        Person.__init__(self, pid, name, role="Engineer")
        self.eng_type = eng_type

    def describe(self):
        return f"{super().describe()}, Eng Type: {self.eng_type}"

class Secretary(Person):
    def __init__(self, pid, name, typing_speed):
        Person.__init__(self, pid, name, role="Secretary")
        self.typing_speed = typing_speed

    def describe(self):
        return f"{super().describe()}, Typing Speed: {self.typing_speed}wpm"

# Disjoint, Total Specialization
def create_specialized_person(pid, name, role, extra):
    if role == "Engineer":
        return Engineer(pid, name, extra)
    elif role == "Secretary":
        return Secretary(pid, name, extra)
    else:
        raise ValueError(f"Error: Total specialization enforced, but no subclass match for {name}")

# Attribute-defined specialization (user-defined)
class Student(Person):
    def __init__(self, pid, name, major):
        Person.__init__(self, pid, name, role="Student")
        self.major = major

    def describe(self):
        return f"{Person.describe(self)}, Major: {self.major}"

class Employee(Person):
    def __init__(self, pid, name, job_type):
        Person.__init__(self, pid, name, role="Employee")
        self.job_type = job_type

    def describe(self):
        return f"{Person.describe(self)}, Job: {self.job_type}"

# Lattice: Multiple inheritance (StudentAssistant subclass of both Student and Employee)
class StudentAssistant(Student, Employee):
    def __init__(self, pid, name, major, job_type):
        # Directly initialize Person to avoid MRO conflicts
        Person.__init__(self, pid, name, role="StudentAssistant")
        self.major = major
        self.job_type = job_type

    def describe(self):
        return (f"{Person.describe(self)} — Roles: "
                f"Student(Major: {self.major}), Employee(Job: {self.job_type})")

# ------------------------------------------------------------------------------
#  Experimentation: Creating instances based on constraints
# ------------------------------------------------------------------------------

print("— Predicate-based Specialization (Disjoint, Total) —")
p1 = create_specialized_person(1, "Alice", "Engineer", "Software")
print(p1.describe())

p2 = create_specialized_person(2, "Bob", "Secretary", 70)
print(p2.describe())

try:
    # Invalid: Total specialization violated (no subclass matches 'Consultant')
    p3 = create_specialized_person(3, "Cathy", "Consultant", None)
    print(p3.describe())
except ValueError as ve:
    print(ve)

# ------------------------------------------------------------------------------
print("\n— User-defined Specialization (Partial, Overlapping) —")

s1 = Student(4, "David", "CS")
print(s1.describe())

e1 = Employee(5, "Eve", "Admin")
print(e1.describe())

# ------------------------------------------------------------------------------
print("\n— Lattice / Multiple Inheritance (StudentAssistant) —")

sa = StudentAssistant(6, "Frank", "Physics", "TA")
print(sa.describe())

# ------------------------------------------------------------------------------
# Optional: modeling combinations explicitly (if multiple inheritance not allowed)
print("\n— Explicit Combination (No Multiple Inheritance Allowed) —")

class StudentEmployeeCombo(Person):
    def __init__(self, pid, name, major, job_type):
        Person.__init__(self, pid, name, role="Student_Employee")
        self.major = major
        self.job_type = job_type

    def describe(self):
        return (f"{Person.describe(self)}, Major: {self.major}, Job: {self.job_type}")

combo = StudentEmployeeCombo(7, "Grace", "Math", "Research Assistant")
print(combo.describe())
```

# 8.4 Modeling of UNION Types Using Categories

---

## 🔹 Definition

- A **category** (also called a **union type**) is a **subclass** that represents a subset of the **union** of two or more **distinct entity types** (superclasses).
- Unlike **shared subclasses** (which are based on **intersections**), **categories** model entities that belong to **only one** of the superclasses.

---

## 🔹 Example

In a **vehicle registration system**, the `OWNER` of a vehicle could be:

- a `PERSON`
- a `BANK`
- a `COMPANY`

Thus:

- `OWNER` is a **category (union type)** subclass of the union of `{PERSON, BANK, COMPANY}`.

---

## 🔹 EER Notation

- Represented by:
  - A **circle** marked with a $\cup$ (**union**) symbol.
  - **Arcs** from each superclass (`BANK`, `COMPANY`, `PERSON`) to the circle.
  - A **subset arc** from the circle to the subclass (e.g., `OWNER`).
- **Defining predicates** may appear next to arcs for clarity.

**Diagram may contain:**
- `OWNER` (from `PERSON`, `BANK`, `COMPANY`)
- `REGISTERED_VEHICLE` (from `CAR`, `TRUCK`)

---

## 🔹 Attribute Inheritance

- An entity in a **category** inherits **only** the attributes of the **specific superclass** it belongs to.

  Example:

  - If `OWNER` is a `PERSON`, it only inherits **PERSON attributes**.
  - Contrast: a **shared subclass** like `ENGINEERING_MANAGER` inherits from **all superclasses**.

---

## 🔹 Category vs. Generalization

- `REGISTERED_VEHICLE` is a **category**:
  - Includes **some** `CAR` and `TRUCK` entities.

- `VEHICLE` in **generalization**:
  - Includes **all** `CAR` and `TRUCK` entities, and possibly more (e.g., `MOTORCYCLE`).

 Categories are more **restrictive** — only entities from **listed superclasses** can be members.

---

## 🔹 Total vs. Partial Categories

- **Total Category**:
  - Includes **all** entities from the union of superclasses.
  - **Diagram**: double line from circle to subclass.

- **Partial Category**:
  - Includes **some** entities from the union.
  - **Diagram**: single line.

---

## 🔹 Key Attributes

- Superclasses in a category may:
  - Have **different keys** (e.g., `OWNER`)
  - Have the **same key** (e.g., `REGISTERED_VEHICLE`)

---

## 🔹 Design Considerations

- Use **specialization/generalization** if:
  - Superclasses **share many attributes** and represent the **same logical type**.

- Use **category** if:
  - Superclasses are **distinct** and have **different attributes**.

Some modeling tools **do not support** union types directly → alternative modeling strategies may be needed.

---


```python
# ---------------------------------------------
# Superclasses: Person, Bank, Company
# ---------------------------------------------

class Person:
    def __init__(self, person_id, name):
        self.person_id = person_id
        self.name = name

    def describe(self):
        return f"PERSON[ID: {self.person_id}, Name: {self.name}]"

class Bank:
    def __init__(self, bank_id, institution_name):
        self.bank_id = bank_id
        self.institution_name = institution_name

    def describe(self):
        return f"BANK[ID: {self.bank_id}, Name: {self.institution_name}]"

class Company:
    def __init__(self, company_id, company_name):
        self.company_id = company_id
        self.company_name = company_name

    def describe(self):
        return f"COMPANY[ID: {self.company_id}, Name: {self.company_name}]"

# ---------------------------------------------
# Category (Union Type): Owner
# ---------------------------------------------

class Owner:
    def __init__(self, entity):
        if isinstance(entity, (Person, Bank, Company)):
            self.entity = entity
        else:
            raise TypeError("Owner must be a Person, Bank, or Company.")

    def describe(self):
        return f"OWNER → {self.entity.describe()}"

# ---------------------------------------------
# RegisteredVehicle as another Category Example
# ---------------------------------------------

class Car:
    def __init__(self, vin, make, model):
        self.vin = vin
        self.make = make
        self.model = model

    def describe(self):
        return f"CAR[VIN: {self.vin}, {self.make} {self.model}]"

class Truck:
    def __init__(self, vin, capacity_tons):
        self.vin = vin
        self.capacity_tons = capacity_tons

    def describe(self):
        return f"TRUCK[VIN: {self.vin}, Capacity: {self.capacity_tons} tons]"

class RegisteredVehicle:
    def __init__(self, vehicle):
        if isinstance(vehicle, (Car, Truck)):
            self.vehicle = vehicle
        else:
            raise TypeError("RegisteredVehicle must be a Car or Truck.")

    def describe(self):
        return f"REGISTERED_VEHICLE → {self.vehicle.describe()}"

# ---------------------------------------------
# Test Cases
# ---------------------------------------------

print("— CATEGORY: OWNER (Union of Person, Bank, Company) —")

p = Person(1, "Alice")
b = Bank(101, "National Bank")
c = Company(501, "TechCorp")

owner1 = Owner(p)
owner2 = Owner(b)
owner3 = Owner(c)

print(owner1.describe())
print(owner2.describe())
print(owner3.describe())

try:
    invalid = Owner("RandomObject")  # Not a valid superclass
except Exception as e:
    print("Error:", e)

print("\n— CATEGORY: REGISTERED_VEHICLE (Union of Car, Truck) —")

car = Car("VIN123", "Toyota", "Corolla")
truck = Truck("VIN789", 10)

reg1 = RegisteredVehicle(car)
reg2 = RegisteredVehicle(truck)

print(reg1.describe())
print(reg2.describe())

try:
    bad = RegisteredVehicle(p)  # Invalid vehicle
except Exception as e:
    print("Error:", e)
```

### 8.5 A Sample UNIVERSITY EER Schema, Design Choices, and Formal Definitions

---

#### 8.5.1 The UNIVERSITY Database Example

A sample EER schema models a **UNIVERSITY** database including:

- **PERSON**: General entity with attributes like `Name`, `SSN`, `Address`, etc.

- **FACULTY** (subclass):  
  Attributes – `Rank`, `Salary`, `Foffice`, `Fphone`.  
  Linked to `DEPARTMENT` (M:N).

- **STUDENT** (subclass):  
  Attribute – `Class (1–5)`.  
  Linked to:
  - `MAJOR`, `MINOR` departments
  - `REGISTERED` in sections
  - `TRANSCRIPT` records past performance

- **GRAD_STUDENT** (subclass of `STUDENT`):  
  Defined by $Class = 5$.  
  Has multivalued attribute `Degrees` and relationships with:
  - `ADVISOR` (FACULTY)
  - `COMMITTEE`

- **DEPARTMENT**:  
  Attributes – `Dname`, `Dphone`, `Office`.  
  Related to `COLLEGE` and has a chair (FACULTY).

- **COLLEGE**:  
  Attributes – `Cname`, `Coffice`, `Dean`.

- **COURSE**:  
  Attributes – `C#`, `Cname`, `Cdesc`.  
  Has multiple `SECTION`s.

- **SECTION**:  
  Attributes – `Sec#`, `Year`, `Qtr`.  
  `CURRENT_SECTION` is a subclass with a time predicate.

- **INSTRUCTOR_RESEARCHER**:  
  A **category (union type)** of `FACULTY` and `GRAD_STUDENT` for those involved in research.

- **GRANT**:  
  Attributes – `Title`, `No`, `Agency`, `St_date`.  
  Linked via:
  - `PI` (Principal Investigator)
  - `SUPPORT` relationship with `% Time`, `Start`, and `End` dates.

---

#### 8.5.2 Design Choices for Specialization/Generalization

**Design Guidelines**:

- Use specialization only when needed to avoid clutter.
- Merge subclasses with few local attributes/relationships into the superclass.
- Use type attributes to distinguish subclasses.
- Avoid union types/categories unless truly necessary.

**Constraints**:

- Use **disjoint** when entities can belong to only one subclass.
- Use **partial** when not all entities must be in a subclass.

**Example**: Instead of creating multiple subclasses, define:

- $Job\_type = \{'Secretary', 'Engineer', 'Technician'\}$
- $Pay\_method = \{'Salaried', 'Hourly'\}$
- $Is\_a\_manager = \{'Yes', 'No'\}$

---

#### 8.5.3 Formal Definitions of EER Concepts

- **Class**: A set of entities (entity type, subclass, category).

- **Subclass $S$ of Superclass $C$**:
  $$ S \subseteq C $$

- **Specialization**: A set of subclasses $\{S_1, S_2, ..., S_n\}$ with a common superclass $G$.
  - **Total**:  
    $$ \bigcup S_i = G $$
  - **Partial**: Not all $G$ entities are in subclasses.
  - **Disjoint**:  
    $$ S_i \cap S_j = \emptyset $$
  - **Overlapping**: Entities can belong to multiple subclasses.

- **Predicate-defined subclass**: Based on a condition (e.g., $Class = 5$).

- **Attribute-defined specialization**: Based on values of an attribute.

- **Category (Union Type) $T$**:  
  A subclass of the union of multiple superclasses:
  $$ T \subseteq (D_1 \cup D_2 \cup \dots \cup D_n) $$

  Optionally defined by predicates:
  $$ T = D_1[p_1] \cup D_2[p_2] \cup \dots \cup D_n[p_n] $$

- **Relationship participation**: Any class (not just entity types) can now participate in relationships.


```python
# -----------------------------------------
# General Entity: PERSON
# -----------------------------------------
class Person:
    def __init__(self, ssn, name, address):
        self.ssn = ssn
        self.name = name
        self.address = address

    def describe(self):
        return f"PERSON[SSN: {self.ssn}, Name: {self.name}, Address: {self.address}]"

# -----------------------------------------
# Subclasses: FACULTY, STUDENT
# -----------------------------------------

class Faculty(Person):
    def __init__(self, ssn, name, address, rank, salary, f_office, f_phone):
        super().__init__(ssn, name, address)
        self.rank = rank
        self.salary = salary
        self.f_office = f_office
        self.f_phone = f_phone
        self.departments = []

    def assign_department(self, dept):
        self.departments.append(dept)

    def describe(self):
        return super().describe() + f" → FACULTY[Rank: {self.rank}, Salary: {self.salary}]"

class Student(Person):
    def __init__(self, ssn, name, address, student_class):
        super().__init__(ssn, name, address)
        self.student_class = student_class
        self.major = None
        self.minor = None

    def set_major_minor(self, major, minor):
        self.major = major
        self.minor = minor

    def describe(self):
        return super().describe() + f" → STUDENT[Class: {self.student_class}]"

# -----------------------------------------
# Predicate-defined Subclass: GRAD_STUDENT
# Class = 5 implies graduate
# -----------------------------------------

class GradStudent(Student):
    def __init__(self, ssn, name, address, degrees=[]):
        super().__init__(ssn, name, address, 5)  # Enforce Class = 5
        self.degrees = degrees
        self.advisors = []
        self.committee = []

    def add_advisor(self, faculty):
        self.advisors.append(faculty)

    def describe(self):
        return super().describe() + f" → GRAD_STUDENT[Degrees: {self.degrees}]"

# -----------------------------------------
# Category (Union): INSTRUCTOR_RESEARCHER
# Subclass of union of FACULTY and GRAD_STUDENT
# -----------------------------------------

class InstructorResearcher:
    def __init__(self, person):
        if isinstance(person, (Faculty, GradStudent)):
            self.person = person
        else:
            raise TypeError("Must be Faculty or GradStudent")

    def describe(self):
        return f"INSTRUCTOR_RESEARCHER → {self.person.describe()}"

# -----------------------------------------
# DEPARTMENT and COLLEGE Entities
# -----------------------------------------

class Department:
    def __init__(self, dname, dphone, office):
        self.dname = dname
        self.dphone = dphone
        self.office = office
        self.chair = None

    def assign_chair(self, faculty):
        self.chair = faculty

    def describe(self):
        return f"DEPT[{self.dname}, Phone: {self.dphone}, Office: {self.office}]"

class College:
    def __init__(self, cname, coffice, dean):
        self.cname = cname
        self.coffice = coffice
        self.dean = dean

    def describe(self):
        return f"COLLEGE[{self.cname}, Office: {self.coffice}, Dean: {self.dean.name}]"

# -----------------------------------------
# COURSE and SECTION
# -----------------------------------------

class Course:
    def __init__(self, code, name, desc):
        self.code = code
        self.name = name
        self.desc = desc
        self.sections = []

    def add_section(self, section):
        self.sections.append(section)

    def describe(self):
        return f"COURSE[{self.code}, {self.name}]"

class Section:
    def __init__(self, sec_no, year, quarter):
        self.sec_no = sec_no
        self.year = year
        self.quarter = quarter

    def describe(self):
        return f"SECTION[{self.sec_no}, {self.year} {self.quarter}]"

# -----------------------------------------
# GRANT and Relationships
# -----------------------------------------

class Grant:
    def __init__(self, title, number, agency, start_date):
        self.title = title
        self.number = number
        self.agency = agency
        self.start_date = start_date
        self.supported_people = []

    def support(self, person, percent_time, start, end):
        self.supported_people.append((person, percent_time, start, end))

    def describe(self):
        return f"GRANT[{self.title}, Agency: {self.agency}, Start: {self.start_date}]"

# -----------------------------------------
# TEST CASES — Demonstrating the Model
# -----------------------------------------

# Person → Faculty + Department
f1 = Faculty("111", "Dr. Smith", "12 Ave", "Professor", 120000, "Room 210", "555-1122")
d1 = Department("CS", "555-CS00", "CS Office")
f1.assign_department(d1)
d1.assign_chair(f1)

# Person → Student (Undergrad) + GradStudent (Class = 5)
s1 = Student("222", "Amy", "45 St", 3)
gs1 = GradStudent("333", "John", "78 St", degrees=["BSc", "MSc"])
gs1.add_advisor(f1)

# Category: InstructorResearcher (FACULTY or GRAD_STUDENT)
ir1 = InstructorResearcher(f1)
ir2 = InstructorResearcher(gs1)

# College
col1 = College("Engineering", "Main Hall", f1)

# Grant Support
g1 = Grant("AI Research", "G-2023", "NSF", "2023-01-01")
g1.support(f1, 50, "2023-01-01", "2024-01-01")

# -----------------------------------------
# OUTPUT
# -----------------------------------------
print(f1.describe())
print(d1.describe())
print(s1.describe())
print(gs1.describe())
print(ir1.describe())
print(ir2.describe())
print(col1.describe())
print(g1.describe())
```

### 8.6 Example of Other Notation: Representing Specialization and Generalization in UML Class Diagrams

---

#### UML Representation of Specialization/Generalization

UML class diagrams are used to represent **inheritance** (specialization/generalization) similar to EER diagrams.

**Notation**:

- Subclasses are connected by vertical lines to a horizontal line, which then connects via a **triangle** to the superclass (base class).
- **Blank triangle**: Disjoint constraint (each object belongs to at most one subclass).
- **Filled triangle**: Overlapping constraint (objects may belong to multiple subclasses).
- **Superclass**: Called the **base class**
- **Subclasses**: Called **leaf classes**

---

#### UML Class Types (Software Engineering Context)

- **Abstract Classes**:
  - Define attributes and operations
  - No objects are instantiated from them
  - Used as a base for inheritance

- **Concrete Classes**:
  - Allow object instantiation
  - Used in database modeling

- **Template Classes**:
  - Serve as generic blueprints
  - Used to define reusable class structures

---

#### Relevance to Database Design

- Focus is primarily on **concrete classes**, as they represent real-world objects stored persistently in databases.
- UML is more detailed in software engineering; database modeling borrows only the necessary constructs for **conceptual schema representation**.


```python
from abc import ABC, abstractmethod

# -----------------------------------------
# UML Abstract Base Class
# -----------------------------------------
class Person(ABC):
    def __init__(self, pid, name):
        self.pid = pid
        self.name = name

    @abstractmethod
    def describe(self):
        pass

# -----------------------------------------
# Disjoint Specialization Classes
# -----------------------------------------
class Student(Person):
    def __init__(self, pid, name, major):
        Person.__init__(self, pid, name)
        self.major = major

    def describe(self):
        return f"Student[ID: {self.pid}, Name: {self.name}, Major: {self.major}]"

class Employee(Person):
    def __init__(self, pid, name, job_type):
        Person.__init__(self, pid, name)
        self.job_type = job_type

    def describe(self):
        return f"Employee[ID: {self.pid}, Name: {self.name}, Job: {self.job_type}]"

# -----------------------------------------
# Overlapping Specialization (Multiple Inheritance)
# -----------------------------------------
class StudentEmployee(Person):
    def __init__(self, pid, name, major, job_type):
        Person.__init__(self, pid, name)
        self.major = major
        self.job_type = job_type

    def describe(self):
        return f"StudentEmployee[ID: {self.pid}, Name: {self.name}, Major: {self.major}, Job: {self.job_type}]"

# -----------------------------------------
# Concrete Non-Inheriting Class
# -----------------------------------------
class Visitor:
    def __init__(self, vid, name, purpose):
        self.vid = vid
        self.name = name
        self.purpose = purpose

    def describe(self):
        return f"Visitor[ID: {self.vid}, Name: {self.name}, Purpose: {self.purpose}]"

# -----------------------------------------
# Test Function
# -----------------------------------------
def test_uml_inheritance():
    print(" UML Disjoint Specialization")
    s1 = Student(1, "Alice", "Physics")
    e1 = Employee(2, "Bob", "Technician")
    print(s1.describe())
    print(e1.describe())

    print("\n UML Overlapping Specialization (Multiple Roles)")
    se1 = StudentEmployee(3, "Charlie", "CS", "RA")
    print(se1.describe())

    print("\n UML Concrete Class (Not abstract)")
    v1 = Visitor(101, "Dana", "Lab Tour")
    print(v1.describe())

    print("\n All classes instantiated and described successfully.")

# Run the test
test_uml_inheritance()
```

### 8.7 Data Abstraction, Knowledge Representation, and Ontology Concepts

---

#### Overview

This section bridges **semantic data modeling** (e.g., EER models) and **Knowledge Representation (KR)** from artificial intelligence. Both use similar concepts like **abstraction**, **classification**, and **relationships**, but differ in scope and application.

---

### 8.7.1 Classification and Instantiation

- **Classification**: Grouping similar entities into classes or entity types.
- **Instantiation**: Creating specific objects (instances) of a class.

**In EER**:
- Entities → Entity types
- Relationship instances → Relationship types

**In UML**:
- Supports class properties and object instances.

**In KR**:
- Supports multiple classification levels, including **meta-classes** (a class can be an instance of another class).

---

### 8.7.2 Identification

Used to uniquely identify:
- Schema constructs (e.g., class names)
- Object instances (e.g., via keys)

**In EER**:
- Every class and attribute must have unique names.
- Key attributes and foreign key references enable identification.
- Needed to relate database tuples to real-world entities they represent.

---

### 8.7.3 Specialization and Generalization

- **Specialization**: Splitting a class into more specific subclasses.
- **Generalization**: Merging multiple classes into a more abstract superclass.

**Representation in EER**:
- Via subclasses and the **IS-A** relationship.
- Supports inheritance, shared attributes, and relationship reuse.

---

### 8.7.4 Aggregation and Association

**Aggregation**:
- Combines component objects into a composite object.
- Example: A `CAR` object composed of `ENGINE`, `TIRES`, etc.

**Representation**:
- Attribute aggregation
- Relationship aggregation
- Composite (molecular) objects – not directly supported in ER but are in some models/UML.

**Association**:
- Connects independent objects through relationships.

**Mappings**:
- EER → Relationship types
- UML → Associations
- Example: `JOB_APPLICANT — interviewed by — COMPANY`

**Key distinction**:
- **Delete association**: Individual objects remain.
- **Delete aggregation**: May delete component objects (if tightly bound).

---

### 8.7.5 Ontologies and the Semantic Web

**Ontology**: A formal specification of a conceptualization of a domain.
- Describes **concepts**, their **relationships**, and **vocabularies**.
- More general and knowledge-rich than database schemas.

**Applications**:
- Used in the **Semantic Web** for intelligent data exchange and automated reasoning.
- Support tools like **search engines**, **recommendation systems**, and **AI reasoning engines**.

---

### Ontology vs Database Schema

| Aspect         | Ontology                          | Database Schema                  |
|----------------|------------------------------------|----------------------------------|
| Scope          | Broader, conceptual                | Narrow, structured for data storage |
| Concepts       | Generalized entities, rules, axioms| Entities, attributes, relationships |
| Usage          | Reasoning, inference, AI           | Data storage and retrieval        |
| Flexibility    | High – includes exceptions         | Limited                           |


```python
# -----------------------------
# 8.7.1 Classification & Instantiation
# -----------------------------
class Person:
    def __init__(self, pid, name):
        self.pid = pid
        self.name = name

    def describe(self):
        return f"Person[ID: {self.pid}, Name: {self.name}]"

# Instantiate (Instantiation)
p1 = Person(1, "Alice")

# -----------------------------
# 8.7.2 Identification (via keys)
# -----------------------------
class Student(Person):
    def __init__(self, pid, name, major):
        super().__init__(pid, name)
        self.major = major

    def describe(self):
        return f"Student[ID: {self.pid}, Name: {self.name}, Major: {self.major}]"

# Unique key pid used to identify student
s1 = Student(2, "Bob", "Physics")

# -----------------------------
# 8.7.3 Specialization & Generalization
# -----------------------------
class Faculty(Person):
    def __init__(self, pid, name, title):
        super().__init__(pid, name)
        self.title = title

    def describe(self):
        return f"Faculty[ID: {self.pid}, Name: {self.name}, Title: {self.title}]"

# General superclass: Person → Specialized: Student, Faculty
f1 = Faculty(3, "Cathy", "Professor")

# -----------------------------
# 8.7.4 Aggregation & Association
# -----------------------------
class Department:
    def __init__(self, dname):
        self.dname = dname
        self.members = []  # Aggregation: group of people

    def add_member(self, person):
        self.members.append(person)  # Aggregation relationship

    def describe(self):
        return f"Department[{self.dname}] Members: {[m.name for m in self.members]}"

# Aggregation
cs_dept = Department("Computer Science")
cs_dept.add_member(s1)
cs_dept.add_member(f1)

# Association: Connect 2 independent classes
class Teaches:
    def __init__(self, faculty, course):
        self.faculty = faculty
        self.course = course

    def describe(self):
        return f"{self.faculty.name} teaches {self.course}"

t1 = Teaches(f1, "Database Systems")

# -----------------------------
# 8.7.5 Ontology-like Conceptualization
# -----------------------------
class Concept:
    def __init__(self, name, description, parent=None):
        self.name = name
        self.description = description
        self.parent = parent

    def describe(self):
        parent_str = f" ⊂ {self.parent.name}" if self.parent else ""
        return f"Concept: {self.name}{parent_str} — {self.description}"

# Simulating a basic ontology
entity = Concept("Entity", "Anything with an identity")
person = Concept("Person", "Human individual", parent=entity)
student = Concept("Student", "A person who is enrolled in a course", parent=person)

# -----------------------------
# Display Results
# -----------------------------
print("\n Classification & Instantiation")
print(p1.describe())
print(s1.describe())

print("\n Specialization & Generalization")
print(f1.describe())

print("\n Aggregation (Department Members)")
print(cs_dept.describe())

print("\n Association (Teaches Relationship)")
print(t1.describe())

print("\n Ontology Representation (Concepts)")
print(entity.describe())
print(person.describe())
print(student.describe())
```

### 8.8 Summary

In this section, we discussed extensions to the ER model that enhance its ability to represent more complex data semantics. These enhancements led to the **Enhanced Entity-Relationship (EER) model**. The key points covered in this chapter are as follows:

---

### Subclass and Superclass Concepts

- Entities may be grouped into **subclasses** for more specific characteristics or relationships.
- Subclasses **inherit attributes and relationships** from their superclasses.

---

### Specialization and Generalization

a. **Specialization**: The process of defining a set of subclasses from a superclass based on distinguishing characteristics.  
b. **Generalization**: The inverse process, where common features from multiple classes are abstracted into a superclass.  
c. These processes form **hierarchies** or **lattices** that help structure complex relationships.

---

### Constraints on Specialization/Generalization

a. **Disjoint vs. Overlapping**: Determines whether an instance can belong to more than one subclass.  
b. **Total vs. Partial**: Specifies whether every instance of a superclass must be a member of some subclass.

---

### Predicate- and Attribute-Defined Specializations

a. **Predicate-defined subclasses**: Use a logical condition to determine class membership.  
b. **Attribute-defined subclasses**: Use specific attribute values to drive classification.

---

### Categories (Union Types)

- Used to represent a **subclass that is a subset of the union** of two or more superclasses.
- Especially useful when these superclasses do **not share a common superclass**.

---

### UML Representation

a. UML uses **triangles** to depict specialization/generalization in class diagrams.  
b. **Blank triangles** indicate disjoint specialization.  
c. **Filled triangles** indicate overlapping.  
d. UML also allows **abstract**, **concrete**, and **template** classes for modeling purposes.

---

### Abstraction and Knowledge Representation (Section 8.7)

a. **Classification and Instantiation**: Classifies objects and relates instances to classes.  
b. **Identification**: Assigns identifiers to distinguish and relate schema objects and real-world entities.  
c. **Specialization and Generalization**: Addressed from both database and KR perspectives.  
d. **Aggregation and Association**: Enables modeling of complex or composite objects; UML supports more forms of aggregation than traditional ER/EER models.

---

### Ontology and Semantic Web

a. **Ontologies** are conceptual frameworks that provide shared vocabulary and relationships to describe a domain.  
b. Used extensively in the **Semantic Web** and **Knowledge Representation** for enabling **machine-readable data** and **intelligent querying**.
